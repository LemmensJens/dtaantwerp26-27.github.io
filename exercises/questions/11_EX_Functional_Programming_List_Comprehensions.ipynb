### 1. Translate the loops (basic)

Rewrite each of the three loops below as one list comprehension, and check that you get the same result.

```python
numbers = [4, 11, 7, 20, 3, 16]

squares = []
for n in numbers:
    squares.append(n * n)

evens = []
for n in numbers:
    if n % 2 == 0:
        evens.append(n)

halved_evens = []
for n in numbers:
    if n % 2 == 0:
        halved_evens.append(n // 2)

print(squares, evens, halved_evens)

# your code here
```
