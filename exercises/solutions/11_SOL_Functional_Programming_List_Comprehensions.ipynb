### 1. Translate the loops (basic)

### 2. Words (basic)

From the sentence below, with one comprehension each: the words in upper case; the words longer than four letters; the length of every word; the first letter of every word, as a single string (join the list).

```python
sentence = 'the quick brown fox jumps over the lazy dog'
words = sentence.split()

print([word.upper() for word in words])
print([word for word in words if len(word) > 4])
print([len(word) for word in words])
print(''.join([word[0] for word in words]))
```

### 3. A corpus of sentences (standard)

Tokenise the three sentences below into one flat list of lowercase tokens without punctuation (use `re.findall(r'[a-z]+', ...)` on each lowercased sentence; a comprehension with two `for` parts, or a loop that extends a list, both work). Then, with comprehensions: the tokens longer than three letters; and the same list with every short token replaced by the string `'_'`.

```python
import re

sentences = ['Alice was beginning to get very tired of sitting by her sister on the bank.',
             'So she was considering in her own mind, as well as she could.',
             'There was nothing so very remarkable in that.']

tokens = [token for sentence in sentences for token in re.findall(r'[a-z]+', sentence.lower())]
print(tokens)

print([token for token in tokens if len(token) > 3])
print([token if len(token) > 3 else '_' for token in tokens])
```

### 4. Positions (standard)

Build a dictionary that maps every position (0, 1, 2, ...) to the word at that position in `words`, with a dictionary comprehension over `enumerate(words)`. Then build the reverse: word to position. Why does the reverse dictionary have fewer entries?

```python
positions = {position: word for position, word in enumerate(words)}
print(positions)

by_word = {word: position for position, word in enumerate(words)}
print(by_word)
# 'the' occurs twice; keys are unique, so the second occurrence overwrites the first.
```

### 5. Sorting novels (standard)

The list below holds (title, author, year) tuples. Sort it by year with a `lambda`; then by the author's surname (the last word of the author's name); then by title length, longest first. Print each result as a list of titles only (a comprehension over the sorted list).

```python
novels = [('Emma', 'Jane Austen', 1815), ('Dracula', 'Bram Stoker', 1897), ('Middlemarch', 'George Eliot', 1871), ('Kidnapped', 'Robert Louis Stevenson', 1886), ('Persuasion', 'Jane Austen', 1817)]

by_year = sorted(novels, key=lambda novel: novel[2])
print([title for title, author, year in by_year])

by_surname = sorted(novels, key=lambda novel: novel[1].split()[-1])
print([title for title, author, year in by_surname])

by_length = sorted(novels, key=lambda novel: len(novel[0]), reverse=True)
print([title for title, author, year in by_length])
```

### 6. A frequency list, sorted (standard)

Read `alice.txt`, tokenise it with `re.findall(r'[a-z]+', text.lower())`, and build a plain dictionary (not a `Counter`) from word to count with a loop. Then print the ten most frequent words by sorting `.items()` with a `lambda` on the count, most frequent first. Check against `Counter(...).most_common(10)`.

```python
with open('../../data/alice.txt', encoding='utf-8') as infile:
    alice = infile.read()

tokens = re.findall(r'[a-z]+', alice.lower())

counts = {}
for token in tokens:
    if token in counts:
        counts[token] += 1
    else:
        counts[token] = 1

top = sorted(counts.items(), key=lambda pair: pair[1], reverse=True)[:10]
print(top)

from collections import Counter
print(Counter(tokens).most_common(10) == top)
```

### 7. Vowel-rich words (standard)

Write a function `count_vowels(word)` and use it as the `key` to sort `tokens` (the *Alice* tokens) by number of vowels, most first. Print the first ten. Then find the single token with the most vowels using `max()` with the same key.

```python
def count_vowels(word):
    return len([char for char in word if char in 'aeiou'])

print(sorted(tokens, key=count_vowels, reverse=True)[:10])
print(max(tokens, key=count_vowels))
```

### 13. Pairs of names (harder)

The function `combinations()` from the module `itertools` gives every pair (or triple, ...) that can be made from a list: `list(combinations(['a', 'b', 'c'], 2))` is `[('a', 'b'), ('a', 'c'), ('b', 'c')]`. For the names below, make the list of all pairs, and then, in a dictionary comprehension over those pairs, count in how many lines of `alice.txt` both names occur (a line counts when `name_1 in line and name_2 in line`). Print the dictionary sorted by count, highest first.

```python
from itertools import combinations

names = ['Alice', 'Queen', 'Hatter', 'Rabbit', 'Gryphon']

pairs = list(combinations(names, 2))
shared = {pair: len([line for line in lines if pair[0] in line and pair[1] in line]) for pair in pairs}

for pair, count in sorted(shared.items(), key=lambda item: item[1], reverse=True):
    print(pair, count)
```

### 15. Exam style: Poirot in comprehensions (harder)

This exercise has the form of an exam question: one file, three steps of increasing difficulty. Read `poirot.txt` and tokenise it with `re.findall(r'[a-z]+', text.lower())`.

(a) *Basic.* With one comprehension, make the list of tokens that are longer than twelve letters, and print how many there are and the first ten.

(b) *Intermediate.* With a dictionary comprehension over `set(long_tokens)`, map every different long token to its number of occurrences (`.count()` on the list of long tokens). Print the five most frequent, sorted with a `lambda`.

(c) *Advanced.* Write a function `average_length(word_list)` that returns the average length of the words in a list, and use it as the `key` to find, with `max()`, which of the three texts `alice.txt`, `poirot.txt` and `Sherlock.txt` has the longest words on average. Read and tokenise each file inside a dictionary comprehension over the three file names, so that you have `{file name: list of tokens}` first.

```python
with open('../../data/poirot.txt', encoding='utf-8') as infile:
    poirot = infile.read()
tokens_poirot = re.findall(r'[a-z]+', poirot.lower())

# (a)
long_tokens = [token for token in tokens_poirot if len(token) > 12]
print(len(long_tokens), long_tokens[:10])

# (b)
long_counts = {token: long_tokens.count(token) for token in set(long_tokens)}
print(sorted(long_counts.items(), key=lambda pair: pair[1], reverse=True)[:5])

# (c)
def average_length(word_list):
    return sum([len(word) for word in word_list]) / len(word_list)

file_names = ['alice.txt', 'poirot.txt', 'Sherlock.txt']
texts = {}
for name in file_names:
    with open('../../data/' + name, encoding='utf-8') as infile:
        texts[name] = re.findall(r'[a-z]+', infile.read().lower())

print({name: round(average_length(words), 2) for name, words in texts.items()})
print(max(texts, key=lambda name: average_length(texts[name])))
```
