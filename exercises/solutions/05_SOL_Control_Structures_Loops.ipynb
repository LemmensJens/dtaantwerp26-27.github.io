### 1. One by one (basic)

Print every title in the list below, one per line, using a `for` loop.

### 2. Which words contain an i? (basic)

For each word in the list, print the word together with `True` or `False` depending on whether it contains the letter *i*.

````python
words = ['This', 'is', 'a', 'list', 'of', 'words']

for word in words:
    print(word, 'i' in word)
````

### 3. Counting characters (basic)

Count how many characters all the words in the list contain together (spaces not included, since they are not in the list) and store the result in `total_characters`. Print it: the answer is 100.

````python
opening = ['In', 'the', 'beginning', 'the', 'Universe', 'was', 'created.', 'This', 'has', 'made', 'a', 'lot', 'of', 'people', 'very', 'angry', 'and', 'been', 'widely', 'regarded', 'as', 'a', 'bad', 'move.']
total_characters = 0

for word in opening:
    total_characters += len(word)

print(total_characters)
````

### 4. Upper case only (basic)

Combine a `for` loop with an `if`: loop over the letters in the list, but only print a letter if it is upper case (hint: the `.isupper()` method).

````python
letters = ['a', 'n', 'Y', 'e', 'm', 'M', 'o', 'C', 'n', 'A', 'e']

for letter in letters:
    if letter.isupper():
        print(letter)
````

### 5. Collecting (standard)

Loop over the letters again, but this time collect all the *lower-case* letters in the empty list `lowercase`. Print the list afterwards to see whether it worked.

````python
letters = ['a', 'n', 'Y', 'e', 'm', 'M', 'o', 'C', 'n', 'A', 'e']
lowercase = []

for letter in letters:
    if letter.islower():
        lowercase.append(letter)

print(lowercase)
````

### 6. Skipping short words (standard)

Write a `for` loop that adds the length of every word to `counter`, but skips words of one or two letters, using `continue`. If all goes to plan, the counter is 40 at the end.

````python
opening = ['I', 'think', 'this', 'is', 'the', 'beginning', 'of', 'a', 'beautiful', 'friendship']
counter = 0

for word in opening:
    if len(word) <= 2:
        continue
    counter += len(word)

print(counter)
````

### 7. A while loop (standard)

Use a `while` loop and the `counter` variable to print the number of letters of every word in the list, one per line. (This is a job for a `for` loop really, but do it with `while` to get the feel.)

````python
counter = 0
sentence = ['I', 'want', 'to', 'fly']

while counter < len(sentence):
    print(len(sentence[counter]))
    counter += 1
````

### 8. Finding the s (standard)

Write a `for` loop that prints a message every time it comes across a word containing the letter *s*. Then rewrite the loop so that it stops after it has found the first such word (hint: `break`).

````python
dwarves = ['happy', 'grumpy', 'sleepy', 'sneezy', 'dopey', 'bashful', 'doc']

for dwarf in dwarves:
    if 's' in dwarf:
        print(f'{dwarf} contains an s')
````

````python
for dwarf in dwarves:
    if 's' in dwarf:
        print(f'{dwarf} is the first dwarf with an s')
        break
````

The list below is a list of lists. For every word that begins with a vowel, print the word and its length, like this:

```

```

Hint: you need a loop inside a loop.

````python
alphabet = [['Alpha', 'Bravo', 'Charlie'], ['Delta', 'Echo', 'Foxtrot'], ['Golf', 'Hotel', 'India', 'Juliet', 'Kilo'], ['Lima', 'Mike', 'November']]

for group in alphabet:
    for word in group:
        if word[0] in 'AEIOU':
            print(word, len(word))
````

### 10. Multiples of five (standard)

Write a `while` loop that prints the counter as long as it is lower than 10, adding 1 after each iteration. Make the loop stop with `break` as soon as the counter hits a multiple of five.

````python
counter = 1

while counter < 10:
    print(counter)
    counter += 1
    if counter % 5 == 0:
        break
````

### 11. Every combination (standard)

Combine every adjective with every noun and print the resulting phrases (`dark night`, `dark forest`, ...). How many phrases do you get?

````python
adjectives = ['dark', 'silent', 'ancient']
nouns = ['night', 'forest', 'castle', 'river']

for adjective in adjectives:
    for noun in nouns:
        print(adjective, noun)

# 3 * 4 = 12 phrases
````

### 12. Skipping empty lines (standard)

Print the poem line by line, but skip the empty lines with `continue`.

````python
poem = ['I wandered lonely as a cloud', 'That floats on high o’er vales and hills,', '', 'When all at once I saw a crowd,', 'A host, of golden daffodils;', '', 'Beside the lake, beneath the trees,', 'Fluttering and dancing in the breeze.']

for line in poem:
    if line == '':
        continue
    print(line)
````

### 13. Reading in instalments (standard)

You read a novel of 450 pages in instalments of 100 pages.

- Create a variable that holds the number of pages read so far (starting at 0).
- Write a loop that adds 100 pages per iteration and prints how many pages you have read.
- When you have read 450 pages or more, print `Finished!` and stop the loop with `break`.

````python
pages_read = 0

while True:
    pages_read += 100
    print(f'{pages_read} pages read')
    if pages_read >= 450:
        print('Finished!')
        break
````

### 14. Average word length (harder)

Turn the paragraph into a list of words with the `.split()` method, then compute the average word length with a loop (no `sum()`). Print the result with two decimals.

````python
paragraph = '''October arrived, spreading a damp chill over the grounds and into the castle.
Madam Pomfrey, the nurse, was kept busy by a sudden spate of colds among the staff and students.
Her Pepperup potion worked instantly, though it left the drinker smoking at the ears for several
hours afterward. Ginny Weasley, who had been looking pale, was bullied into taking some by Percy.
The steam pouring from under her vivid hair gave the impression that her whole head was on fire.'''

words = paragraph.split()
total_length = 0

for word in words:
    total_length += len(word)

average = total_length / len(words)
print(f'{average:.2f}')
````

### 16. Titles and years (harder)

The two lists belong together: the first title was published in the first year, and so on. Use `zip()` to walk through them in parallel (`for title, year in zip(titles, years):`) and print a sentence for each pair. Then, in a second loop, print only the titles published before 1850.

````python
titles = ['Emma', 'Dracula', 'Persuasion', 'Middlemarch', 'Kidnapped']
years = [1815, 1897, 1817, 1871, 1886]

for title, year in zip(titles, years):
    print(f'{title} was published in {year}.')

for title, year in zip(titles, years):
    if year < 1850:
        print(title)
````
