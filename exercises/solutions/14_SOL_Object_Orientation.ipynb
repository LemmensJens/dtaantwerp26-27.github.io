### Class exercise 1: predict the output

```python
words = ['pear', 'apple', 'fig']
title = 'moby dick'

a = sorted(words)
b = words.sort()
c = title.upper()
d = words.append('kiwi')

print(a)         # ['apple', 'fig', 'pear']: sorted() returns a new list
print(b)         # None: .sort() sorts in place and returns nothing
print(words)     # ['apple', 'fig', 'pear', 'kiwi']: sorted in place, then 'kiwi' appended
print(c)         # MOBY DICK: .upper() returns a new string
print(title)     # moby dick: strings are never changed in place
print(d)         # None: .append() changes the list and returns nothing
```

### Class exercise 2: `total()` and `remove()`

```python
aeneid = """Arma virumque cano, Troiae qui primus ab oris
Italiam fato profugus Lavinaque venit
litora multum ille et terris iactatus et alto
vi superum, saevae memorem Iunonis ob iram,
multa quoque et bello passus, dum conderet urbem
inferretque deos Latio; genus unde Latinum
Albanique patres atque altae moenia Romae."""


class OurCounter:

    def __init__(self, iterable=None):
        self.counts = {}
        if iterable:
            for item in iterable:
                self.add(item)

    def add(self, key, amount=1):
        try:
            self.counts[key] += amount
        except KeyError:
            self.counts[key] = amount

    def most_common(self, n=None):
        pairs = sorted(self.counts.items(), key=lambda pair: pair[1], reverse=True)
        return pairs[:n]

    def total(self):
        return sum(self.counts.values())

    def remove(self, key):
        if key in self.counts:
            del self.counts[key]

    def __str__(self):
        return 'OurCounter(' + str(self.counts) + ')'


cnt = OurCounter(aeneid)
print(cnt.total(), len(aeneid))
cnt.remove(' ')
cnt.remove('not there')
print(cnt.total())
print(cnt.most_common(3))
```

### Easy: Counter skills on Alice

```python
from collections import Counter

with open('../../data/alice.txt', encoding='utf-8') as infile:
    text = infile.read()

print(Counter(text).most_common(3))
print(Counter(text.lower().split()).most_common(3))

line_counters = [Counter(line) for line in text.splitlines()]
print(len(line_counters))

richest = max(line_counters, key=len)
print(len(richest), richest)
```

### Advanced: a small library

```python
class Book:

    def __init__(self, title, shelfmark, year):
        self.title = title
        self.shelfmark = shelfmark
        self.year = year

    def __str__(self):
        return f'{self.title} ({self.year}), shelfmark {self.shelfmark}'


class User:

    def __init__(self, name):
        self.name = name
        self.loans = []

    def list_loans(self):
        if not self.loans:
            print(f'{self.name} has no books on loan.')
            return
        print(f'{self.name} has on loan:')
        for book in self.loans:
            print('  ' + str(book))


class StaffMember:

    def __init__(self, name):
        self.name = name

    def lend(self, book, user):
        if book in user.loans:
            print(f'{user.name} already has {book.title}.')
            return
        user.loans.append(book)
        print(f'{self.name} lends {book.title} to {user.name}.')

    def take_back(self, book, user):
        if book not in user.loans:
            print(f'{user.name} does not have {book.title}.')
            return
        user.loans.remove(book)
        print(f'{self.name} takes {book.title} back from {user.name}.')


emma = Book('Emma', 'ENG 823 AUS', 1815)
dracula = Book('Dracula', 'ENG 823 STO', 1897)
ada = User('Ada')
librarian = StaffMember('Mr Dodgson')

print(emma)
librarian.lend(emma, ada)
librarian.lend(dracula, ada)
librarian.lend(emma, ada)
ada.list_loans()
librarian.take_back(emma, ada)
librarian.take_back(emma, ada)
ada.list_loans()
```
