### 5. First and last (standard)

Write a function `same_edges(word)` that returns `True` if a word starts and ends with the same letter, ignoring case, and `False` otherwise; one-letter words do not count. Then print the words of the excerpt from exercise 4 for which it is `True`, after stripping commas and full stops from each word with `.strip('.,')` (the answer is *that* and *pop*). Use `.startswith()` and `.endswith()`, or indexing; both are fine.

```python
def same_edges(word):
    word = word.lower()
    return len(word) > 1 and word.startswith(word[-1])

for word in excerpt.split():
    word = word.strip('.,')
    if same_edges(word):
        print(word)
```

### 8. Tokenise and count (standard)

Write `tokenize(text)` as in the session (lowercase, split, strip punctuation, drop empty strings) and use it on `alice.txt`. Print the number of tokens (28129), the number of different tokens (3149), and how often *queen* and *rabbit* occur (68 and 43). Use a `Counter`.

```python
import string
from collections import Counter

def tokenize(text):
    tokens = []
    for piece in text.lower().split():
        token = piece.strip(string.punctuation)
        if token:
            tokens.append(token)
    return tokens

with open('../../data/alice.txt', encoding='utf-8') as infile:
    alice = infile.read()

tokens = tokenize(alice)
counts = Counter(tokens)
print(len(tokens), len(counts))
print(counts['queen'], counts['rabbit'])
```

### 10. Capitalise every word (standard)

Write a function `capitalise_words(text)` that capitalises the first letter of every word and leaves the rest of each word alone, *without* using `.title()` or `.capitalize()`: split, change each word with slicing, join. Test it on `'do androids dream of electric SHEEP?'` (the result keeps `SHEEP?` as it is).

```python
def capitalise_words(text):
    words = []
    for word in text.split():
        words.append(word[0].upper() + word[1:])
    return ' '.join(words)

print(capitalise_words('do androids dream of electric SHEEP?'))
```

### 11. Swap the case (standard)

Write `swap_case(text)` that turns upper case into lower case and vice versa, without `.swapcase()`: loop over the characters, collect the changed ones in a list, join. Test on `'Do Androids Dream?'`.

```python
def swap_case(text):
    chars = []
    for char in text:
        if char.isupper():
            chars.append(char.lower())
        else:
            chars.append(char.upper())
    return ''.join(chars)

print(swap_case('Do Androids Dream?'))
```

### 12. Remove a character (standard)

Write `remove_at(text, n)` that returns `text` without its character at index `n`, using slicing only. `remove_at('Hello, World!', 7)` gives `'Hello, orld!'`. What should happen when `n` is beyond the end of the string? Decide, and make the function do it.

```python
def remove_at(text, n):
    if n >= len(text):
        raise ValueError(f'index {n} is beyond the end of the text (length {len(text)})')
    return text[:n] + text[n + 1:]

print(remove_at('Hello, World!', 7))
```

### 13. Only vowels (harder)

Write `vowels_only(text)` that returns the text with every consonant removed, keeping vowels, spaces and punctuation. `vowels_only('Curiouser and curiouser!')` gives `'uioue a uioue!'`. Then `consonants_only()`, the other way round.

```python
def vowels_only(text):
    kept = []
    for char in text:
        if not char.isalpha() or char.lower() in 'aeiou':
            kept.append(char)
    return ''.join(kept)

def consonants_only(text):
    kept = []
    for char in text:
        if not char.isalpha() or char.lower() not in 'aeiou':
            kept.append(char)
    return ''.join(kept)

print(vowels_only('Curiouser and curiouser!'))
print(consonants_only('Curiouser and curiouser!'))
```

### 15. Exam style: a song (harder)

This exercise has the form of an exam question: one file, three steps of increasing difficulty. The file `MT.txt` contains the lyrics of a song.

(a) *Basic.* Read the file and print its number of lines (28) and how often the string `ritz` occurs, ignoring case (11).

(b) *Intermediate.* Write `first_words(text)` that returns a list with the first word of every non-empty line, cleaned with `.strip(string.punctuation)` and lowercased. Print the result.

(c) *Advanced.* Tokenise the song with your `tokenize()` from exercise 8 and print the five most frequent tokens. Then write the song to a new file `ritz_clean.txt` in which every line is replaced by its cleaned tokens joined with single spaces, one line per original line.

```python
import string
from collections import Counter

# (a)
with open('../../data/MT.txt', encoding='utf-8') as infile:
    song = infile.read()
lines = song.splitlines()
print(len(lines))
print(song.lower().count('ritz'))

# (b)
def first_words(text):
    result = []
    for line in text.splitlines():
        if line.strip():
            result.append(line.split()[0].strip(string.punctuation).lower())
    return result

print(first_words(song))

# (c)
print(Counter(tokenize(song)).most_common(5))

with open('ritz_clean.txt', 'w', encoding='utf-8') as outfile:
    for line in lines:
        outfile.write(' '.join(tokenize(line)) + '\n')
```

**Initials.** `initials('Charles Lutwidge Dodgson')` gives `'C.L.D.'`.

```python
def initials(name):
    letters = []
    for part in name.split():
        letters.append(part[0].upper())
    return '.'.join(letters) + '.'

print(initials('Charles Lutwidge Dodgson'))
print(initials('lewis carroll'))
```

**Sentences, roughly.** Splitting on `'. '` misses sentences that end in `?` or `!` (and a full stop at the very end has no space after it, so the last piece keeps its stop). It would also split at abbreviations such as *Mr.* in other texts. The session on regular expressions gives a better tool.

```python
paragraph = """Alice was beginning to get very tired of sitting by her sister on the bank, and of having nothing to do:  once or twice she had peeped into the book her sister was reading, but it had no pictures or conversations in it, `and what is the use of a book,' thought Alice `without pictures or conversation?' So she was considering in her own mind (as well as she could, for the hot day made her feel very sleepy and stupid), whether the pleasure of making a daisy-chain would be worth the trouble of getting up and picking the daisies, when suddenly a White Rabbit with pink eyes ran close by her. There was nothing so VERY remarkable in that; nor did Alice think it so VERY much out of the way to hear the Rabbit say to itself, `Oh dear!  Oh dear!  I shall be late!'"""

for number, sentence in enumerate(paragraph.split('. '), 1):
    print(number, sentence.strip())
```
