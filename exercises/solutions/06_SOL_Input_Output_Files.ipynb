### 6. Lines (standard)

### 8. Save and reload (standard)

Write the reading list below to a file `reading_list.txt`, one novel per line, with a tab between the title and the year. Then read the file back in and print every line without its line break. Open the file in a text editor too, to see what it looks like.

```python
titles = ['Emma', 'Dracula', 'Persuasion', 'Middlemarch', 'Kidnapped']
years = [1815, 1897, 1817, 1871, 1886]

with open('reading_list.txt', 'w', encoding='utf-8') as outfile:
    for title, year in zip(titles, years):
        outfile.write(f'{title}\t{year}\n')

with open('reading_list.txt', encoding='utf-8') as infile:
    for line in infile:
        print(line.strip())
```

### 9. Sherlock speaks first (standard)

Loop over the lines of `Sherlock.txt` *without* reading the whole file first (loop over the open file object). Write every line whose first word is `Sherlock` to a new file `sherlock_lines.txt`, and print how many lines you wrote: the answer is 32. One thing to watch: the first word of a line is `line.split()[0]`, but an empty line has no first word, so skip empty lines.

```python
written = 0

with open('../../data/Sherlock.txt', encoding='utf-8') as infile:
    with open('sherlock_lines.txt', 'w', encoding='utf-8') as outfile:
        for line in infile:
            if line.strip() and line.split()[0] == 'Sherlock':
                outfile.write(line)
                written += 1

print(written)
```

### 10. The longest line (harder)

Find the longest line in `alice.txt` (after stripping its whitespace) and print it together with its length. Keep track of the longest line seen so far in a variable, and replace it whenever you meet a longer one. The winner has 75 characters.

```python
longest = ''

with open('../../data/alice.txt', encoding='utf-8') as infile:
    for line in infile:
        line = line.strip()
        if len(line) > len(longest):
            longest = line

print(longest)
print(len(longest))
```

### 11. A log that grows (standard)

Open a file `log.txt` in append mode and write three lines to it in a loop: `Entry 1`, `Entry 2`, `Entry 3`. Then read the file back and print the number of lines in it. Run the cell a second time: how many lines are there now, and why? Write the explanation as a comment in your cell.

```python
with open('log.txt', 'a', encoding='utf-8') as outfile:
    for number in range(1, 4):
        outfile.write(f'Entry {number}\n')

with open('log.txt', encoding='utf-8') as infile:
    lines = infile.readlines()

print(len(lines))
# The second run prints 6: append mode keeps what is in the file and adds to it,
# so every run adds three more lines. Write mode ('w') would start from scratch each time.
```

### 12. Empty lines, proportionally (harder)

Which text has the higher share of empty lines, `alice.txt` or `poirot.txt`? Loop over the list of paths, and for each file compute the percentage of empty lines, rounded to one decimal, and print it together with the path, as in `../../data/alice.txt: 24.0%`. (`round(x, 1)` rounds to one decimal.)

```python
paths = ['../../data/alice.txt', '../../data/poirot.txt']

for path in paths:
    with open(path, encoding='utf-8') as infile:
        lines = infile.readlines()
    empty = 0
    for line in lines:
        if not line.strip():
            empty += 1
    percentage = round(empty / len(lines) * 100, 1)
    print(f'{path}: {percentage}%')
```

### 13. Rock, paper, scissors (harder)

Write a game of rock, paper, scissors against the computer. `random.choice(some_list)` picks a random element from a list: use it for the computer's move. Ask the user for a move and keep asking until it is one of `rock`, `paper` or `scissors`. Then print both moves and the outcome: rock beats scissors, scissors beat paper, paper beats rock, and equal moves are a tie.

```python
import random

moves = ['rock', 'paper', 'scissors']
computer = random.choice(moves)

while True:
    user = input('Rock, paper or scissors? ')
    if user in moves:
        break
    print('Please type rock, paper or scissors.')

print(f'You: {user}. Computer: {computer}.')

if user == computer:
    print('A tie.')
elif (user == 'rock' and computer == 'scissors') or (user == 'scissors' and computer == 'paper') or (user == 'paper' and computer == 'rock'):
    print('You win!')
else:
    print('The computer wins.')
```

### 14. Chapters (harder)

The chapter headings in `alice.txt` are lines that start with `CHAPTER` once you strip the whitespace around them (`line.strip()[:7] == 'CHAPTER'`). Collect them in a list, write them to `chapters.txt` (one per line), and print how many there are: 12.

```python
chapters = []
with open('../../data/alice.txt', encoding='utf-8') as infile:
    for line in infile:
        if line.strip()[:7] == 'CHAPTER':
            chapters.append(line.strip())

with open('chapters.txt', 'w', encoding='utf-8') as outfile:
    for chapter in chapters:
        outfile.write(chapter + '\n')

print(len(chapters))
```

### 15. Exam style: Puttin' on the Ritz (harder)

This exercise has the form of an exam question: one file, three steps of increasing difficulty. The file `MT.txt` contains the lyrics of a song.

(a) *Basic.* Read the file as a list of lines and print how many lines and how many characters it contains (28 lines; count the characters of the whole text as one string).

(b) *Intermediate.* Write a numbered copy of the song to `ritz_numbered.txt`: every line is preceded by its number (starting at 1) and a tab. Use `enumerate()` from the session on loops.

(c) *Advanced.* Count the lines that contain the word `ritz`, ignoring case, and write only those lines to `ritz_lines.txt`. Print the count: it is 9.

```python
# (a)
with open('../../data/MT.txt', encoding='utf-8') as infile:
    text = infile.read()
lines = text.splitlines()
print(len(lines), len(text))

# (b)
with open('ritz_numbered.txt', 'w', encoding='utf-8') as outfile:
    for number, line in enumerate(lines, 1):
        outfile.write(f'{number}\t{line}\n')

# (c)
count = 0
with open('ritz_lines.txt', 'w', encoding='utf-8') as outfile:
    for line in lines:
        if 'ritz' in line.lower():
            outfile.write(line + '\n')
            count += 1
print(count)
```
