### 4. Letters (standard)

Build a frequency dictionary of the **letters** in the excerpt below (the preface of *The Picture of Dorian Gray*). Count upper and lower case together, and count only letters (`.isalpha()`). Print the dictionary. How often does the letter *t* occur? The answer is 65.

```python
dorian = """The artist is the creator of beautiful things. To reveal art and
conceal the artist is art’s aim. The critic is he who can translate
into another manner or a new material his impression of beautiful
things.

The highest as the lowest form of criticism is a mode of autobiography.
Those who find ugly meanings in beautiful things are corrupt without
being charming. This is a fault.

Those who find beautiful meanings in beautiful things are the
cultivated. For these there is hope. They are the elect to whom
beautiful things mean only beauty.

There is no such thing as a moral or an immoral book. Books are well
written, or badly written. That is all.

The nineteenth century dislike of realism is the rage of Caliban seeing
his own face in a glass."""

letters = {}
for char in dorian.lower():
    if char.isalpha():
        if char in letters:
            letters[char] += 1
        else:
            letters[char] = 1

print(letters)
print(letters['t'])
```

### 5. Words (standard)

Build a frequency dictionary of the **words** in the Dorian Gray excerpt: lowercase the text and split it on whitespace. How many different words are there (83), and how often does *beautiful* occur (6)?

```python
word_counts = {}
for word in dorian.lower().split():
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

print(len(word_counts))
print(word_counts['beautiful'])
```

### 6. The most frequent word (standard)

Using the dictionary of the previous exercise, find the most frequent word and its count with a loop that keeps track of the best pair seen so far (as in the session). The answer is *the*, 10 times.

```python
most_frequent = ''
highest_count = 0

for word, count in word_counts.items():
    if count > highest_count:
        most_frequent = word
        highest_count = count

print(most_frequent, highest_count)
```

### 7. Numbers with gaps (standard)

The list below contains page counts as strings, but some entries are not numbers. Convert the entries to integers and collect them in `pages`, skipping the ones that cannot be converted, with `try`/`except`. Print the list and the sum of the pages.

```python
raw_pages = ['320', '150', 'unknown', '410', '', '275', 'n/a']
pages = []

for raw in raw_pages:
    try:
        pages.append(int(raw))
    except ValueError:
        print(f'skipping "{raw}"')

print(pages)
print(sum(pages))
```

### 8. Counter (standard)

Read `poirot.txt` from the data folder, lowercase it, split it into words and count them with a `Counter`. Print the ten most common words, and how often *poirot* occurs (309; the tokens `poirot,` and `poirot.` are counted separately, which is a limitation of splitting on whitespace).

```python
from collections import Counter

with open('../../data/poirot.txt', encoding='utf-8') as infile:
    poirot = infile.read()

poirot_counts = Counter(poirot.lower().split())

print(poirot_counts.most_common(10))
print(poirot_counts['poirot'])
```

### 13. Who appears with whom (harder)

For every line of `alice.txt`, find which of the names in the list below occur in that line, and count how often each **pair** of names shares a line. Use a dictionary whose keys are tuples of two names, such as `('Alice', 'Queen')`. For every line, loop over the names that are present and, for every name after it in that list, add one to the count of the pair. Print the pairs with their counts. *Alice* and the *Gryphon* share 7 lines.

```python
names = ['Alice', 'Queen', 'Hatter', 'Rabbit', 'Duchess', 'Gryphon']

pair_counts = {}
with open('../../data/alice.txt', encoding='utf-8') as infile:
    for line in infile:
        present = []
        for name in names:
            if name in line:
                present.append(name)
        for i in range(len(present)):
            for j in range(i + 1, len(present)):
                pair = (present[i], present[j])
                if pair in pair_counts:
                    pair_counts[pair] += 1
                else:
                    pair_counts[pair] = 1

for pair, count in pair_counts.items():
    print(pair, count)
```

### 14. A small table (harder)

The header and the rows below describe a few novels. Turn every row into a dictionary with `dict(zip(header, row))` and collect these in a list `table`. Then loop over the table and print the titles of the novels published before 1850.

```python
header = ['title', 'author', 'year']
rows = [['Emma', 'Jane Austen', 1815],
        ['Dracula', 'Bram Stoker', 1897],
        ['Persuasion', 'Jane Austen', 1817],
        ['Middlemarch', 'George Eliot', 1871]]

table = []
for row in rows:
    table.append(dict(zip(header, row)))

for novel in table:
    if novel['year'] < 1850:
        print(novel['title'])
```

### 15. Sorting by value (harder)

A dictionary has no order of its own, and `sorted(word_counts)` sorts its **keys**. What you usually want from a frequency dictionary is the opposite: the entries sorted by their **value**, most frequent first. That is less straightforward than you would expect from a modern programming language: there is no `.sort_by_value()`. The classic trick works with what you already know. Build a list of tuples with the count *first* and the word second, `(count, word)`, one per entry, by looping over `.items()`. Python sorts a list of tuples by the first element of each tuple (and by the second when the first elements are equal), so `sorted(pairs, reverse=True)` puts the highest count first (`reverse=True` reverses the order). Apply this to the `word_counts` dictionary of the Dorian Gray excerpt and print the five most frequent words with their counts, one per line, most frequent first.

(`Counter.most_common()` does exactly this for you, which is why you will normally use it. The general solution, `sorted(..., key=...)`, comes in the session on functional programming.)

```python
pairs = []
for word, count in word_counts.items():
    pairs.append((count, word))

for count, word in sorted(pairs, reverse=True)[:5]:
    print(word, count)
```

### 16. Exam style: Poirot investigates (harder)

This exercise has the form of an exam question: one file, three steps of increasing difficulty.

(a) *Basic.* Read `poirot.txt`, lowercase it and split it into words. Print the number of words (52501).

(b) *Intermediate.* Build a frequency dictionary of the words with a loop (no `Counter`), and print how many words occur exactly once.

(c) *Advanced.* With a `Counter`, find the ten most frequent words that consist of letters only (`.isalpha()`) and are longer than six letters. Build the list of such words first, then count.

```python
# (a)
with open('../../data/poirot.txt', encoding='utf-8') as infile:
    poirot = infile.read()
words = poirot.lower().split()
print(len(words))

# (b)
counts = {}
for word in words:
    if word in counts:
        counts[word] += 1
    else:
        counts[word] = 1

once = 0
for word, count in counts.items():
    if count == 1:
        once += 1
print(once)

# (c)
from collections import Counter

long_words = []
for word in words:
    if word.isalpha() and len(word) > 6:
        long_words.append(word)

print(Counter(long_words).most_common(10))
```
