### 1. Count the characters (basic)

Write a function `count_characters(text)` that returns the number of characters in `text` *without spaces*. Test it on `'Alice in Wonderland'` (17) and on an empty string (0).

```python
def count_characters(text):
    count = 0
    for char in text:
        if char != ' ':
            count += 1
    return count

print(count_characters('Alice in Wonderland'))
print(count_characters(''))
```

### 3. Long words (standard)

Write a function `long_words(text, min_length=6)` that returns a list of the words in `text` that are longer than `min_length` characters. Call it on the sentence below once with the default and once with `min_length=4`, and once more with a keyword argument.

```python
sentence = 'Alice was beginning to get very tired of sitting by her sister on the bank'

def long_words(text, min_length=6):
    result = []
    for word in text.split():
        if len(word) > min_length:
            result.append(word)
    return result

print(long_words(sentence))
print(long_words(sentence, 4))
print(long_words(sentence, min_length=3))
```

### 4. Describe a novel (standard)

Write a function `describe(title, author, year)` that returns a sentence such as `Emma, by Jane Austen, was published in 1815.` Call it once with positional arguments and once with keyword arguments in a different order. Then give `year` a default value of `None` and make the function return `Emma, by Jane Austen.` when no year is given (test with `if year is None:`; `is` is the way to test for `None`).

```python
def describe(title, author, year=None):
    if year is None:
        return f'{title}, by {author}.'
    return f'{title}, by {author}, was published in {year}.'

print(describe('Emma', 'Jane Austen', 1815))
print(describe(year=1897, author='Bram Stoker', title='Dracula'))
print(describe('Persuasion', 'Jane Austen'))
```

### 5. Check the input (standard)

Write a function `average_word_length(text)` that returns the average length of the words in `text`, rounded to two decimals. If `text` is not a string, the function should `raise` a `TypeError` with a clear message; if it contains no words at all, a `ValueError`. Test it on the sentence from exercise 3 (4.0), and show that a list and an empty string produce the errors.

```python
def average_word_length(text):
    if not isinstance(text, str):
        raise TypeError(f'expected a string, got {type(text)}')
    words = text.split()
    if len(words) == 0:
        raise ValueError('the text contains no words')
    total = 0
    for word in words:
        total += len(word)
    return round(total / len(words), 2)

print(average_word_length(sentence))
```

### 6. Two results (standard)

Write a function `min_and_max(numbers)` that returns both the smallest and the largest number in a list, as a tuple, *without* using `min()` or `max()`: keep the smallest and largest seen so far while you loop. Unpack the result into two variables and print them. Test on `[4, 11, 2, 9, 7]` (2 and 11).

```python
def min_and_max(numbers):
    smallest = numbers[0]
    largest = numbers[0]
    for number in numbers:
        if number < smallest:
            smallest = number
        if number > largest:
            largest = number
    return smallest, largest

lowest, highest = min_and_max([4, 11, 2, 9, 7])
print(lowest, highest)
```

```python
def acrostic(poem):
    """Return the first letters of the lines of poem (a string with one line per verse) as one string."""
    letters = ''
    for line in poem.splitlines():
        letters += line[0]
    return letters

help(acrostic)
```

### 8. Decomposition (harder)

Preprocess the three sentences from *Alice* below in three steps, each its own function: `lowercase(sentence)` returns the lowercased sentence; `tokenize(sentence)` returns the list of words (split on whitespace); `preprocess(sentence_list)` applies both to every sentence and returns a list of lists of words. Test the two small functions on the first sentence before you write the third. Finally, print the number of words in each preprocessed sentence (15, 24 and 35).

```python
sentences = ["Alice was beginning to get very tired of sitting by her sister on the bank.",
             "So she was considering in her own mind (as well as she could, for the hot day made her feel very sleepy and stupid).",
             "There was nothing so VERY remarkable in that; nor did Alice think it so VERY much out of the way to hear the Rabbit say to itself, 'Oh dear! Oh dear! I shall be late!'"]

def lowercase(sentence):
    return sentence.lower()

def tokenize(sentence):
    return sentence.split()

def preprocess(sentence_list):
    result = []
    for sentence in sentence_list:
        result.append(tokenize(lowercase(sentence)))
    return result

print(lowercase(sentences[0]))
print(tokenize(sentences[0]))

for words in preprocess(sentences):
    print(len(words), words[:5])
```

### 10. Palindromes (standard)

Write a function `is_palindrome(text)` that returns `True` if `text` reads the same backwards as forwards, ignoring case and spaces, and `False` otherwise. Two helpers: the string method `text.replace(' ', '')` returns a copy of `text` with every space removed (more on `.replace()` in the session on strings), and slicing with a step of `-1` (`text[::-1]`) reverses a string. Test on `'Level'`, `'Was it a car or a cat I saw'` and `'Alice'`.

```python
def is_palindrome(text):
    cleaned = text.lower().replace(' ', '')
    return cleaned == cleaned[::-1]

print(is_palindrome('Level'))
print(is_palindrome('Was it a car or a cat I saw'))
print(is_palindrome('Alice'))
```

### 11. Your own len() (standard)

Write a function `my_len(sequence)` that returns the number of elements in a sequence without using `len()`. Test it on a string, a list and a dictionary (for a dictionary, the number of keys).

```python
def my_len(sequence):
    count = 0
    for element in sequence:
        count += 1
    return count

print(my_len('Wonderland'))
print(my_len(['this', 'is', 'a', 'list']))
print(my_len({'Emma': 1815, 'Dracula': 1897}))
```

### 12. The list trap (harder)

Read the cell below and predict what the two `print()` calls show. Run it. Then fix `alphabetical()` so that it returns a sorted copy and leaves the original list untouched (the session showed how; `sorted()` is another way), and check that `words` is unchanged afterwards.

```python
# Both prints show ['a', 'cat', 'mat', 'on', 'the']: .sort() sorted the very list that words
# refers to, because word_list and words are two names for one list.

def alphabetical(word_list):
    result = word_list.copy()
    result.sort()
    return result

words = ['mat', 'cat', 'the', 'on', 'a']
sorted_words = alphabetical(words)

print(sorted_words)
print(words)
```

```python
# Alternative
def alphabetical(word_list):
    return sorted(word_list)

print(alphabetical(words))
print(words)
```

### 13. Random lines (standard)

The `random` module has a function `random.choice(items)` that returns one random element of a list, and `random.sample(items, k)` that returns `k` different random elements as a list. Import the module under the alias `rd` and write a function `random_lines(path, k=1)` that reads the file at `path`, keeps the lines that are not empty, and returns `k` of them at random, stripped of their line breaks. Call it on `alice.txt` with the default and with `k=3`.

```python
import random as rd

def random_lines(path, k=1):
    lines = []
    with open(path, encoding='utf-8') as infile:
        for line in infile:
            if line.strip():
                lines.append(line.strip())
    return rd.sample(lines, k)

print(random_lines('../../data/alice.txt'))
print(random_lines('../../data/alice.txt', k=3))
```

### 14. Vowels, consonants and the rest (harder)

Write a function `character_counts(text)` that returns three numbers as a tuple: the number of vowels, the number of consonants and the number of other characters (punctuation, digits, ...) in `text`, ignoring spaces and case. Decompose: first write `is_vowel(char)` and `is_consonant(char)` (a consonant is a letter that is not a vowel; `.isalpha()` tests for letters). Give the main function a docstring. Test on the first sentence of exercise 8 (21 vowels, 39 consonants, 1 other).

```python
def is_vowel(char):
    return char.lower() in 'aeiou'

def is_consonant(char):
    return char.isalpha() and not is_vowel(char)

def character_counts(text):
    """Return (vowels, consonants, other) for the characters in text, spaces not counted."""
    vowels = 0
    consonants = 0
    other = 0
    for char in text:
        if char == ' ':
            continue
        if is_vowel(char):
            vowels += 1
        elif is_consonant(char):
            consonants += 1
        else:
            other += 1
    return vowels, consonants, other

print(character_counts(sentences[0]))
```

### 15. Exam style: a small text toolkit (harder)

This exercise has the form of an exam question: three functions of increasing difficulty that build on each other.

(a) *Basic.* Write `read_text(path)` that reads the file at `path` and returns its contents as one lowercased string. Test it on `poirot.txt` (295958 characters).

(b) *Intermediate.* Write `count_word(text, word)` that returns how often `word` occurs as a separate word in `text` (split on whitespace; do not worry about punctuation). How often does `poirot` occur? (309)

(c) *Advanced.* Write `most_frequent_long_words(text, n=10, min_length=7)` that returns the `n` most frequent words in `text` consisting of letters only and having at least `min_length` characters, as a list of (word, count) tuples, using a `Counter`. Print the result for the Poirot text with the defaults, and the top five words of at least ten letters.

```python
from collections import Counter

# (a)
def read_text(path):
    with open(path, encoding='utf-8') as infile:
        return infile.read().lower()

poirot = read_text('../../data/poirot.txt')
print(len(poirot))

# (b)
def count_word(text, word):
    count = 0
    for token in text.split():
        if token == word:
            count += 1
    return count

print(count_word(poirot, 'poirot'))

# (c)
def most_frequent_long_words(text, n=10, min_length=7):
    long_words = []
    for token in text.split():
        if token.isalpha() and len(token) >= min_length:
            long_words.append(token)
    return Counter(long_words).most_common(n)

print(most_frequent_long_words(poirot))
print(most_frequent_long_words(poirot, n=5, min_length=10))
```

**Decomposition.** The ten most frequent clean words in *Alice*.

```python
from collections import Counter

def clean_word(word):
    """Lowercase word and keep only its letters."""
    cleaned = ''
    for char in word.lower():
        if char.isalpha():
            cleaned += char
    return cleaned

def tokenize(text):
    """Split text on whitespace and clean every word; drop words that are empty after cleaning."""
    tokens = []
    for word in text.split():
        cleaned = clean_word(word)
        if cleaned:
            tokens.append(cleaned)
    return tokens

def most_frequent(path, n=10):
    """Return the n most frequent clean words in the file at path."""
    with open(path, encoding='utf-8') as infile:
        text = infile.read()
    return Counter(tokenize(text)).most_common(n)

print(clean_word("Alice's"), clean_word('(said)'))
print(most_frequent('../../data/alice.txt'))
```
