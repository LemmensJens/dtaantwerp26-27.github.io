## Core exercises

### 1. Warming up (basic)

Use `re.findall()` on the string below to find: (a) every *i*; (b) every *a* or *o*; (c) every vowel; (d) every digit; (e) every capital letter; (f) every character that is not a letter, a digit or a space.

```python
import re

song = "My nr. 1 childhood song is 'Supercalifragilisticexpialidocious'!!!"

print(re.findall(r'i', song))
print(re.findall(r'[ao]', song))
print(re.findall(r'[aeiou]', song))
print(re.findall(r'\d', song))
print(re.findall(r'[A-Z]', song))
print(re.findall(r'[^a-zA-Z0-9 ]', song))
```

### 2. Raw or not (basic)

Predict what the two calls below return, then run the cell. Explain the difference in a comment.

```python
song = "My nr. 1 childhood song is 'Supercalifragilisticexpialidocious'!!!"

print(re.findall('\\bsong\\b', song))
print(re.findall(r'\bsong\b', song))

# The first pattern is a normal string, so Python turns \b into a backspace character before
# re sees it, and the pattern 'backspace song backspace' matches nothing. In the raw string
# the backslashes reach re untouched, and \b means word boundary.
```

### 3. Words, by shape (standard)

On the excerpt from *Dracula* below, find with one `re.findall()` each: every word of eight letters or more; every capitalised word; every word that ends in *s*, as a whole word (use `\b`).

```python
stoker = '''In the population of Transylvania there are four distinct nationalities:
Saxons in the South, and mixed with them the Wallachs, who are the
descendants of the Dacians; Magyars in the West, and Szekelys in the
East and North. I am going among the latter, who claim to be descended
from Attila and the Huns. This may be so, for when the Magyars conquered
the country in the eleventh century they found the Huns settled in it. I
read that every known superstition in the world is gathered into the
horseshoe of the Carpathians, as if it were the centre of some sort of
imaginative whirlpool; if so my stay may be very interesting.'''

print(re.findall(r'\b\w{8,}\b', stoker))
print(re.findall(r'\b[A-Z]\w*', stoker))
print(re.findall(r'\b\w+s\b', stoker))
```

### 4. Where is it? (standard)

Write a function `find_span(pattern, text)` that returns the `(start, end)` span of the *first* match of `pattern` in `text`, or `None` if there is no match. Which function of `re` is the right one? Test with the calls below (the last one should give `None`).

```python
def find_span(pattern, text):
    match = re.search(pattern, text)
    if match:
        return match.span()
    return None

print(find_span(r'Magyars', stoker))
print(find_span(r'\d+', stoker))
print(find_span(r'Dracula', stoker))
```

### 5. Cleaning (standard)

Turn the messy string below into a clean list of lowercase words with `re.sub()` and `.split()`: remove the digits, remove everything that is not a word character or whitespace, collapse runs of whitespace. Expected: `['chapter', 'the', 'pool', 'of', 'tears', 'page']`.

```python
messy = 'CHAPTER 2:   The  Pool\tof Tears...   (page 23)'

cleaned = re.sub(r'\d+', '', messy)
cleaned = re.sub(r'[^\w\s]', '', cleaned)
print(cleaned.lower().split())
```

### 6. Sentences (standard)

Split the Dracula excerpt into sentences with `re.split()` (a full stop, exclamation or question mark followed by whitespace), and print every sentence with its number and its number of words. How many sentences are there? (4)

```python
sentences = re.split(r'[.!?]\s+', stoker)
print(len(sentences))
for number, sentence in enumerate(sentences, 1):
    print(number, len(sentence.split()), sentence[:50])
```

### 7. Titles and years (standard)

The string below lists novels with their year between brackets. With one `re.findall()` and two capturing groups, extract every (title, year) pair; `re.findall()` returns a list of tuples when the pattern has more than one group. Then build a dictionary from title to year (as an integer) and print it.

```python
catalogue = 'Emma (1815), Dracula (1897), Middlemarch (1871), Kidnapped (1886), Persuasion (1817)'

pairs = re.findall(r'(\w+) \((\d{4})\)', catalogue)
print(pairs)

years = {}
for title, year in pairs:
    years[title] = int(year)
print(years)
```

### 8. Adverbs (standard)

Read `alice.txt`, lowercase it and find every word that ends in *-ly* as a whole word. Print the five most frequent with a `Counter` (the first is *only*, 52 times).

```python
from collections import Counter

with open('../../data/alice.txt', encoding='utf-8') as infile:
    alice = infile.read()

ly_words = re.findall(r'\b\w+ly\b', alice.lower())
print(Counter(ly_words).most_common(5))
```

Find the three e-mail addresses in the text below. An address, roughly: one or more word characters or dots, an `@`, one or more word characters, a dot, and two or more letters. Build the pattern in three steps and test each.

Find every different word in `alice.txt` (lowercased) that contains three vowels in a row, and print them sorted. (Use a set to remove duplicates.)

### 11. Hyphens (standard)

Find every hyphenated word in `alice.txt` (two runs of word characters with a hyphen between them) and print the eight most common. Some hits are not really words: which, and why does the pattern find them?

```python
hyphenated = re.findall(r'\b\w+-\w+\b', alice)
print(Counter(hyphenated).most_common(8))
# 'bread-and' (from bread-and-butter) and 'out-of' / 'the-way' (from out-of-the-way): the pattern
# takes exactly two parts, so a word with three parts is split into overlapping pieces.
# 'GUTENBERG-tm' comes from the Project Gutenberg licence text, not from the novel.
```

### 12. The word before (harder)

In the Dracula excerpt, find every word that is followed by a word starting with *a* (upper or lower case), allowing an optional comma or semicolon between them. Use a capturing group for the word you want and leave the following word outside the group. Expected: `['there', 'South', 'who', 'West', 'East', 'I', 'going', 'from', 'Carpathians']`. One word is missing from that list although it *is* followed by *and*: which one, and why does the pattern miss it? (Hint: matches never overlap; *from Attila and* is the place to look.)

```python
print(re.findall(r'(\w+)[,;]?\s+[aA]\w*', stoker))
# 'Attila' is missing: the match 'from Attila' has already consumed the word Attila, and the
# next match can only start after it, so 'Attila and' is never seen.
```

### 13. Line ends (harder)

Find the last word of every line of the Dracula excerpt: a run of word characters followed by a line break. Then do the same for the *first* word of every line (a line break followed by word characters; the very first line has no line break before it, so handle it separately or use `^` with the flag `re.MULTILINE`, which makes `^` match at the start of every line).

```python
print(re.findall(r'(\w+)\n', stoker))
print(re.findall(r'^(\w+)', stoker, re.MULTILINE))
```

### 14. Shouting (harder)

The Gutenberg text of *Alice* uses capitals for emphasis (*VERY*). Find every word of two or more capital letters, as a whole word, and print the ten most common. Some of them come from the Gutenberg header rather than from the novel: which?

### 15. Exam style: Poirot investigates (harder)

This exercise has the form of an exam question: one file, three steps of increasing difficulty. Read `poirot.txt` into a string first.

(a) *Basic.* How often does *Poirot* occur as a whole word (use `\b`)? Compare with `.count('Poirot')`. The two are equal here; write in a comment when they would differ.

(b) *Intermediate.* Find every title followed by a capitalised name (*Monsieur*, *Madame*, *Mademoiselle*, *Mr*, *Mrs* or *Miss*, an optional full stop, a space, a capitalised word) with two capturing groups, and print the ten most common (title, name) pairs with a `Counter`.

(c) *Advanced.* Split the text into sentences with `re.split()` on `.`, `!` or `?` followed by whitespace. Print the number of sentences and the average number of words per sentence, rounded to one decimal (count words with `re.findall(r'\w+', sentence)`).

```python
with open('../../data/poirot.txt', encoding='utf-8') as infile:
    poirot = infile.read()

# (a)
print(len(re.findall(r'\bPoirot\b', poirot)))
print(poirot.count('Poirot'))
# They would differ if 'Poirot' occurred inside a longer word (as 'cat' does in 'catalogue'):
# .count() would count that, \bPoirot\b would not. Poirot's still counts for both, because the
# apostrophe is not a word character, so there is a word boundary after Poirot.

# (b)
titled = re.findall(r'\b(Monsieur|Madame|Mademoiselle|Mr|Mrs|Miss)\.? ([A-Z]\w+)', poirot)
print(Counter(titled).most_common(10))

# (c)
sentences = re.split(r'[.!?]\s+', poirot)
total_words = 0
for sentence in sentences:
    total_words += len(re.findall(r'\w+', sentence))
print(len(sentences))
print(round(total_words / len(sentences), 1))
```
