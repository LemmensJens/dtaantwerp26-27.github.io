### 9. Wholesale price (standard)

The cover price of a book is 24.95 euros, but bookshops get a 40 percent discount. Shipping costs 3 euros for the first copy and 75 cents for each additional copy. What is the total cost of an order of 60 copies? Print the result in a readable sentence, using `str()` where necessary.

```python
cover_price = 24.95
discount = 0.40
copies = 60

price_per_copy = cover_price * (1 - discount)
shipping = 3 + 0.75 * (copies - 1)
total = copies * price_per_copy + shipping

print('The order costs ' + str(total) + ' euros.')
```

The result has a long tail of decimals (`945.4499999999999`): computers store decimal numbers with a tiny imprecision. This is normal, and later we will see how to round numbers when printing them.

### 11. How long is a novel? (standard)

A novel has 320 pages; a page has on average 32 lines, and a line about 11 words. Estimate the number of words in the novel in one line of code, store the result in a variable with a clear name, and print it. Then estimate how many hours it takes to read the novel at 250 words per minute.

```python
pages = 320
lines_per_page = 32
words_per_line = 11

number_of_words = pages * lines_per_page * words_per_line
print(number_of_words)

reading_hours = number_of_words / 250 / 60
print(reading_hours)
```

### 12. Swap (standard)

Write code that swaps the values of the two variables below, so that `first_number` becomes 5 and `second_number` becomes 3. Hint: you need an extra variable.

```python
first_number = 3
second_number = 5

temporary = first_number
first_number = second_number
second_number = temporary

print(first_number, second_number)
```

Without the extra variable, the first assignment would overwrite one of the values before you have copied it. (Python has a shortcut for this, `first_number, second_number = second_number, first_number`, which you will meet later.)

### 15. Leftover books (harder)

Ten books are divided evenly among three students. How many books are left over? Look up online what the `%` operator ("modulo") does and use it. Bonus: solve the same problem *without* `%`, using `//` (integer division) instead, and check your solution with a different number of books.

```python
books = 10
students = 3

leftover = books % students
print(leftover)
```

```python
# Bonus: without the modulo operator
leftover = books - (books // students) * students
print(leftover)
```

`%` gives the remainder of a division; `//` gives the result of the division rounded down to a whole number. Both come back in the next session.
