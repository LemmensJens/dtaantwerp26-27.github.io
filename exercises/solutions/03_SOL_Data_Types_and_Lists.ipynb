### 1. What type is it? (basic)

What are the data types of the variables below? Write down your guess for each one, then check with the `type()` function.

````python
a = False
b = ('Hello', 'there')
c = 24601
d = 3.14
e = ['Hello', False, 24601, 3.14]
f = 5 + 1.0
g = 15 - 20

print(type(a))   # bool
print(type(b))   # tuple (round brackets)
print(type(c))   # int
print(type(d))   # float
print(type(e))   # list (square brackets); the elements have different types, which is fine
print(type(f))   # float: int + float gives a float
print(type(g))   # int: negative numbers are integers too
````

### 7. Welcome (standard)

`Welcome to Python club "PHILEAS FOGG" We are happy to have you with us!`

Hint: use the `.upper()` method for the name.

````python
first_name = 'Phileas'
last_name = 'Fogg'
part_1 = 'Welcome to Python club'
part_2 = 'We are happy to have you with us!'

name = first_name.upper() + ' ' + last_name.upper()
greeting = part_1 + ' "' + name + '" ' + part_2
print(greeting)
````

### 14. Relative frequency (harder)

Compute the *relative* frequency of a given character in a text: how often it occurs, divided by the total number of characters. Count upper and lower case together, and print the result with three decimals (an f-string with `:.3f`).

````python
text = '''And for this reason, I said, money and honour have no attraction for them; good men do not wish to be openly demanding payment for governing and so to get the name of hirelings, nor by secretly helping themselves out of the public revenues to get the name of thieves. And not being ambitious they do not care about honour. Wherefore necessity must be laid upon them, and they must be induced to serve from the fear of punishment. And this, as I imagine, is the reason why the forwardness to take office, instead of waiting to be compelled, has been deemed dishonourable. Now the worst part of the punishment is that he who refuses to rule is liable to be ruled by one who is worse than himself.'''
character = 'a'

count = text.lower().count(character)
relative_frequency = count / len(text)
print(f'The character {character} makes up {relative_frequency:.3f} of the text.')
````

### 17. A reading list (harder)

Below are three reading lists. Perform the following operations and print the result after each step:

1. add `'Frankenstein'` to `gothic`;
2. remove `'The Monk'` from `gothic`;
3. find the index of `'Persuasion'` in `austen` (hint: the `.index()` method);
4. replace `'Persuasion'` in `austen` with `'Northanger Abbey'`, using that index;
5. sort `austen` alphabetically;
6. print every second title of `victorian`;
7. reverse `victorian`;
8. make one list `reading_lists` that contains the three lists (a list of lists);
9. print the first two titles of the second list in `reading_lists`.

````python
gothic = ['The Castle of Otranto', 'The Monk', 'Dracula', 'Carmilla']
austen = ['Emma', 'Persuasion', 'Pride and Prejudice', 'Mansfield Park']
victorian = ['Middlemarch', 'Jane Eyre', 'Bleak House', 'Vanity Fair', 'North and South', 'Cranford']

# 1
gothic.append('Frankenstein')
print(gothic)
# 2
gothic.remove('The Monk')
print(gothic)
# 3
position = austen.index('Persuasion')
print(position)
# 4
austen[position] = 'Northanger Abbey'
print(austen)
# 5
austen.sort()
print(austen)
# 6
print(victorian[::2])
# 7
victorian.reverse()
print(victorian)
# 8
reading_lists = [gothic, austen, victorian]
print(reading_lists)
# 9
print(reading_lists[1][:2])
````
