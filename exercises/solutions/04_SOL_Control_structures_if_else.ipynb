### 7. Ratings (standard)

A review site turns scores into ratings. Print:

- `A` if the score is higher than 90;
- `B` if it is higher than 60;
- `C` if it is higher than 50;
- `F` in all other cases.

Use `if`, `elif` and `else`, and make sure the tests do not overlap.

```python
score = 75

if score > 90:
    print('A')
elif score > 60:
    print('B')
elif score > 50:
    print('C')
else:
    print('F')
```

The order of the tests matters: `score > 60` would also be true for 95, but a score of 95 never reaches that test because the first one already caught it.

### 11. Seasons (standard)

Ask the user for a season with `input()` and print the date on which that season starts:

- spring starts on 20 March;
- summer starts on 21 June;
- autumn starts on 23 September;
- winter starts on 21 December.

Accept `fall` as well as `autumn` (hint: `or`), and make the code work whatever capitalisation the user types (hint: `.lower()`). Print a message for anything that is not a season.

```python
season = input('Which season? ').lower()

if season == 'spring':
    print('Spring starts on 20 March.')
elif season == 'summer':
    print('Summer starts on 21 June.')
elif season == 'autumn' or season == 'fall':
    print('Autumn starts on 23 September.')
elif season == 'winter':
    print('Winter starts on 21 December.')
else:
    print('That is not a season.')
```

### 12. Literary periods (standard)

Print the period a novel belongs to, based on its year of publication:

- before 1660: early modern;
- from 1660 to 1789: Restoration and eighteenth century;
- from 1790 to 1836: Romantic;
- from 1837 to 1900: Victorian;
- after 1900: modern.

Make sure that every year ends up in exactly one period. Test with a few different years.

```python
year = 1851

if year < 1660:
    print('early modern')
elif year <= 1789:
    print('Restoration and eighteenth century')
elif year <= 1836:
    print('Romantic')
elif year <= 1900:
    print('Victorian')
else:
    print('modern')
```

Because each `elif` only sees the years the previous tests did not catch, the lower bound of every period is implied: `year <= 1789` is only tested for years of 1660 and later.

### 14. Divisible (harder)

Write code that prints whether the number is:

- divisible by both two and three;
- divisible by two only;
- divisible by three only;
- divisible by neither.

Exactly one message should be printed. Think about the order of the tests.

```python
number = 12

if number % 2 == 0 and number % 3 == 0:
    print('divisible by two and three')
elif number % 2 == 0:
    print('divisible by two only')
elif number % 3 == 0:
    print('divisible by three only')
else:
    print('divisible by neither')
```

The combined test has to come first: if `number % 2 == 0` were tested first, 12 would be caught there and never reach the combined test.

### 15. Only the first (harder)

The cell prints two lines. Change the *structure* of the code (not the numbers) so that only the first statement is printed.

```python
number = 6

if number < 9:
    print('This should be printed')
elif number > 18:
    print('This should not be printed')
else:
    print('This should not be printed either')

# The two tests were independent if statements, so the second one ran (and its else fired).
# With elif, the second test belongs to the first, and is skipped once the first succeeds.
```

### 16. The shortest word (harder)

Print the shortest of the three words, using only `if`, `elif`, `else` and comparisons (no `min()`). What does your code do when two words are equally long?

```python
word_1 = 'serendipity'
word_2 = 'ink'
word_3 = 'quill'

if len(word_1) <= len(word_2) and len(word_1) <= len(word_3):
    print(word_1)
elif len(word_2) <= len(word_3):
    print(word_2)
else:
    print(word_3)

# With <=, the first of two equally short words wins; with < it would be the last one.
```
