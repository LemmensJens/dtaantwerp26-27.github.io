An essential programming cliché is the **loop**.

Suppose you want to take a list of words as input, and print out the length of each word of that list.

## The `for` loop

Luckily, there are more efficient ways to do this in Python.

Whenever you see a problem where the same thing (here: computing a length) has to be done for every element of a list, a string or any other sequence, you should think **`for` loop**. Let's do the previous problem as a loop.

The syntax of a `for` loop is easy: you start with the keyword `for`, followed by a variable name of your own choice, followed by the keyword `in`, followed by the input sequence, terminated by a colon (`:`).

```python
for w in ['cheese', 'and', 'onions']:
```
The name of the variable (`w`) is your own choice; it will be bound to each word in the list in turn, until the end of the list is reached.

Then, in an indented block (just like with `if`), we specify what to do in each **iteration**, with `w` bound to the current word. This block is executed as many times as there are elements in the sequence.

```python
print(len(w))
```

Remember that `w` will be bound to `'cheese'`, `'and'` and `'onions'` in turn, and the loop stops after that.

So, a `for` loop does something for each element in a sequence. Let's see what this gives:

````python
for word in ['cheese', 'and', 'onions']:
    print(len(word))
````

A few more examples and use cases. A `for` loop works on any sequence, so also on strings (where it walks through the characters) and on ranges:

````python
for letter in "Wolf":
    print(letter)
````

````python
for x in ['A', 'b', 'c', 'D']:
    print(x.islower())   # str.islower() checks whether a string is in lowercase
````

````python
list_of_numbers = range(4, 10, 2)    # all integers starting with 4 up to 10 (not included) in steps of 2
for number in list_of_numbers:
    print(number)
````

A very useful `for` cliché is to do something for each element of a sequence, **accumulate** the results in a variable, and do something with that variable afterwards. For example, compute the average length of the words in a list:

````python
title = ["The", "Physics", "of", "Climate", "Change"]

cumulated_length = 0

for w in title:   
    cumulated_length += len(w)   # Recall: a += 1 is the same as a = a + 1

print(cumulated_length / len(title))
````

Sometimes it is useful to **trace** how the values of variables change in each iteration of the loop. A simple way to do that is to add a `print()` inside the loop (with an f-string, so that you see which iteration you are in):

````python
title = ["The", "Physics", "of", "Climate", "Change"]

cumulated_length = 0

for w in title:   
    cumulated_length += len(w)   # Recall: a += 1 is the same as a = a + 1
    print(f'cumulated_length after word {w} is {cumulated_length}') # here the variables are being placed into the string

print(cumulated_length / len(title))
````

Another useful idiom is to combine `for` with a test, to create a **filter** on a list: `for` loop + `if` = filter.

````python
# Filter all even numbers from the list of integers from 1 to 100

result = []
for n in range(1, 101):  # why 101?
    if n % 2 == 0:   # an even number has 0 remainder when divided by 2
        result.append(n)
print(result)
````

## Two helpers: `enumerate()` and `zip()`

Two functions come up so often in loops that you should know them from the start.

Often you need the **position** of an element as well as the element itself, for instance to number the lines of a poem. `enumerate()` gives you both: in each iteration it hands you a pair (position, element), which you catch with *two* loop variables.

````python
lines = ['Shall I compare thee to a summer’s day?', 'Thou art more lovely and more temperate:']

for number, line in enumerate(lines):
    print(number, line)
````

Remember that Python counts from zero; `enumerate(lines, 1)` starts the numbering at 1.

When you have **two sequences that belong together**, element by element, `zip()` walks through them in parallel:

````python
titles = ['Emma', 'Dracula', 'Middlemarch']
years = [1815, 1897, 1871]

for title, year in zip(titles, years):
    print(f'{title} was published in {year}.')
````

The pairs that `enumerate()` and `zip()` hand out are **tuples** (the immutable sequences from the session on data types), and writing two variable names in the `for` line is called **unpacking** the tuple. You will see the same trick when we loop over dictionaries.

## The `while` loop

Another way to implement an iteration is the `while` loop which, conceptually, is a combination of an `if` and a `for` loop. The syntax is very similar to an `if` block: a Python keyword (`while`) is followed by a test. After that, a colon follows and an indented code block:

```python
while a < 6:
    print(a)
```

Like with `if`, the associated code block is executed only if the test holds. The catch, however, is that the block keeps on getting executed, over and over, until the test no longer holds. In this, `while` resembles `for`: both are iterative.

**Danger:** `while` loops can be a nightmare, because they keep going until the test returns `False`. If that never happens, you are stuck in an endless loop.

- If you can solve the problem with a `for` loop, use that. Every `for` loop can be written as a `while` loop, but not the other way round, so there are cases where you need a `while`.
- Make sure that the test becomes `False` at some point, before you run the cell.
- If an endless loop does happen, use *Kernel > Interrupt* (or the stop button) in Jupyter. In a terminal, `Ctrl+C` does the same.

````python
counter = 1 
while counter < 11:
    print(counter)
    counter += 1 
````

We initialize a variable with value `1`. Starting the `while` loop, `counter < 11` will return `True`, so the code block after the test will be executed: the current value of `counter` is printed and its value is incremented by one. By incrementing the counter in each iteration, we know the test will at some point return `False`, and the execution will stop, i.e., we will jump out of the `while` loop and continue with what follows if there is anything following.

## `break`

A useful way to make sure that a `while` loop ends is `break`, another Python keyword. You can insert it inside the indented block of a loop, and it halts the loop immediately.

````python
counter = 0

while True: # recipe for an endless loop
    counter += 1
    print(counter)
    if counter == 10:
        break # saved by the break
````

- The `break` keyword is **nested** inside an `if` statement that is itself part of a `while` loop. Python stops the nearest enclosing loop, which matters when loops are nested.
- With `while True:` the loop is intentionally set up to be infinite: the test can never become `False`, so make sure you have a `break` in that case.

`break` can also be used in `for` loops, with the same effect: it terminates the loop immediately. Have a look at this example, in which we search through a sentence until we find the first occurrence of the word "sleep".

````python
sentence = ['colourless', 'green', 'ideas', 'sleep', 'furiously']

for word in sentence:
    print(word)
    if word == 'sleep':
        print("found it!")
        break
    
````

We iterate over the words in the sentence and when we come across "sleep", we halt the loop. Note that the last word doesn't get printed anymore: we skip the rest of the loop because we've already found what we were looking for. This is a classic use of `break`: avoiding unnecessary work.

Watch out with nesting: `break` only stops a single loop, the nearest one that encloses it. Suppose that we want to search our sentence for the first occurrence of the letter *d*:

````python
for word in sentence: # outer, sentence-level for-loop
    print('')
    for letter in word: # inner, word-level for-loop
        print(letter, end=' ') # variant of print without the implicit \n
        if letter == 'd':
            print("\nFound a 'd'!")
            break
````

The `break` only halts the inner loop (so that the rest of the current word, "ideas", isn't searched anymore after the *d*). The outer loop happily keeps on running. You typically need a `break` per loop in such situations.

## `continue`

The keyword `break` has a sibling: `continue`. It works in `for` loops as well as `while` loops: when Python encounters it, it skips the rest of the *current iteration* and moves on to the next one. A simple illustration:

````python
word = 'alphabet'
for letter in word:
    if letter == 'a':
        continue # move on to the next iteration
    print(letter)
````

All letters are printed except the *a*: encountering the `continue` takes you straight to the start of the *next* iteration. Let's look at a slightly bigger application. Suppose that we have the following consonants and vowels:

````python
vowels = 'aeiou'
consonants = 'ptkbdg'
````

How can we generate all consonant-vowel combinations? 

````python
for v in vowels:
    for c in consonants:
        print(c, v, sep='', end=' ')   # sep: what goes between the printed items (default: a space); end: what comes after (default: a newline)
````

How could we extend this to a CVC skeleton?

````python
for o in consonants: # o for onset
    for v in vowels: 
        for c in consonants: # c for coda
            print(o, v, c, sep='', end=' ')
````

Now, let's keep only the cases where the two consonants are the same (words like "dod" or "bab"). Can we use `continue` for this?

````python
for o in consonants:
    for v in vowels:
        for c in consonants:
            if o != c:
                continue   # skip to the next iteration when the consonants differ
            print(o, v, c, sep='', end=' ')
````

Could we use `break` here? Why (not)?

Remember:
- A `break` exits from the smallest / nearest enclosing `for` or `while` loop.
- A `continue` exits *the current iteration* of the smallest / nearest enclosing `for` or `while` loop.


````python
for o in consonants:
    for v in vowels:
        for c in consonants:
            if o != c:
                break   # gives up on the whole inner loop as soon as one coda differs: not what we want
            print(o, v, c, sep='', end=' ')
````

## Optional: `while` with `else`

A `while` loop (and a `for` loop) can have an `else` block, which runs only when the loop ended *normally*, i.e. not through a `break`. You will rarely need it, but you may come across it, so here is what it looks like: the message `No results found.` is only printed if the letter was not found.

````python
word = 'colourless'
query = 'r' # also try 'x'
counter = 0

while counter < len(word):
    letter = word[counter]
    print(letter)
    if letter == query:
        print('Found one!')
        break
    counter += 1
else:
    print('No results found.')
````
