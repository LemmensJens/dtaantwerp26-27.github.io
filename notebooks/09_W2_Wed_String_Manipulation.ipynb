## Strings cannot be changed

A string is **immutable**: once created, it cannot be altered. Indexing lets you *read* a character, not overwrite it:

```python
# This cell produces an error on purpose: strings do not support item assignment
word = 'mouse'
word[0] = 'h'
```

This has one consequence that catches everyone at least once. A string method never changes the string it is called on; it **returns a new string**, and if you do not store that new string, it is lost:

```python
title = 'moby dick'
title.upper()          # a new string is created and thrown away
print(title)

title = title.upper()  # store the result: this is the pattern
print(title)
```

Compare this with lists, where `.append()` and `.sort()` change the list in place and return `None`. With strings it is always `variable = variable.method()`.

## Special characters: escape sequences and raw strings

Some characters cannot simply be typed inside quotes: a line break, a tab, a quotation mark of the same kind as the one that delimits the string. Python writes them with a backslash, as **escape sequences**. You have used `\n` since the session on files; here is the short list:

| Sequence | Meaning |
| --- | --- |
| `\n` | newline |
| `\t` | tab |
| `\'` and `\"` | a quote that does not end the string |
| `\\` | one backslash |

The backslash is a signal to Python, not a character in the string: `len('\n')` is 1.

```python
print('First line\nSecond line')
print('Name:\tAlice')
print('She said \'hello\' and left.')
print('A backslash: \\')
print(len('\n'))
```

Sometimes you want a backslash to be *just a backslash*, for instance in a Windows path or, from the next session on, in a regular expression pattern, where backslashes are everywhere. A **raw string**, written with an `r` before the opening quote, switches escape sequences off: every character means itself.

```python
print('C:\new\table')        # \n and \t are interpreted
print(r'C:\new\table')       # raw string: the backslashes stay
```

Remember the `r`: in the session on regular expressions you will write every pattern as `r'...'`.

## Building strings: f-strings

To build a string out of pieces, you know `+`, and you know its limitation: everything has to be a string already, so numbers need `str()` first.

```python
# This cell produces an error on purpose: you cannot add a number to a string
title = 'Emma'
year = 1815
print(title + ' was published in ' + year)
```

**f-strings** are the standard way to do this, and you have used them since the session on data types: an `f` before the quote, and any expression between curly brackets is evaluated and inserted, whatever its type.

```python
print(f'{title} was published in {year}.')
print(f'{title} has {len(title)} letters and was published {2026 - year} years ago.')
print(f'{title.upper()}!')
```

Two extras that are worth knowing. After a colon inside the brackets you can add a **format specification**: `.2f` means "as a number with two decimals", which is what you want for averages and percentages; a number such as `>10` reserves a width of ten characters and aligns the value to the right, which lines up columns when you print a table.

```python
ratio = 5473 / 28200
print(f'ratio: {ratio}')
print(f'ratio: {ratio:.2f}')
print(f'percentage: {ratio * 100:.1f}%')

for title, year in [('Emma', 1815), ('Dracula', 1897), ('Middlemarch', 1871)]:
    print(f'{title:<12}{year:>6}')      # < left-aligned in 12 characters, > right-aligned in 6
```

You will also meet an older method in other people's code, `'{} was published in {}'.format(title, year)`. It does the same as an f-string with the values listed afterwards; read it when you see it, but write f-strings.

## Searching and counting

Four tools tell you whether and where something occurs in a string. `in` answers yes or no; `.count()` counts non-overlapping occurrences; `.find()` gives the index of the first occurrence, or `-1` if there is none; `.startswith()` and `.endswith()` test the beginning and end.

```python
sentence = 'The Queen of Hearts, she made some tarts, all on a summer day.'

print('tarts' in sentence)
print(sentence.count('a'))
print(sentence.find('Hearts'))         # index of the first character of the match
print(sentence.find('Alice'))          # not found
print(sentence.startswith('The'))
print(sentence.endswith('.'))
print(sentence.lower().startswith('the queen'))
```

All of these are **case sensitive**: `'queen' in sentence` is `False`. When case does not matter, lowercase the string first, as in the last line. `.startswith()` and `.endswith()` are the methods you want for filtering lines and file names (`line.startswith('CHAPTER')`, `filename.endswith('.txt')`); they are much clearer than comparing slices.

There is also `.index()`, which does what `.find()` does but raises a `ValueError` when the substring is absent. Prefer `.find()`, or test with `in` first.

## Changing strings

Since strings cannot be changed, "changing" always means "making a changed copy". The workhorse is `.replace(old, new)`, which replaces *every* occurrence:

```python
sentence = 'The Queen of Hearts, she made some tarts.'

print(sentence.replace('Queen', 'King'))
print(sentence.replace(' ', '_'))
print(sentence.replace(',', ''))            # replacing with nothing: deleting
print(sentence.replace('e', 'E', 2))        # optional third argument: at most this many replacements
```

### Case

`.lower()` and `.upper()` you know. `.title()` capitalises every word, `.capitalize()` only the first character of the string and lowercases the rest. There is no method for "capitalise the first letter and leave the rest alone", but slicing gives it to you in one line, an idiom worth remembering:

```python
text = 'the ADVENTURES of sherlock holmes'

print(text.title())
print(text.capitalize())
print(text[0].upper() + text[1:])       # first letter up, rest untouched
```

### Whitespace and punctuation at the edges

`.strip()` removes whitespace (spaces, tabs, newlines) from both ends; `.lstrip()` and `.rstrip()` from one end. With an argument, `.strip()` removes any of the *characters you give it* instead of whitespace, which is how you take the punctuation off a word:

```python
word = '"tarts,"'

print(word.strip('",'))
print('...well...'.strip('.'))
print('   indented\n'.strip())
```

Typing every punctuation mark by hand is error-prone. The module `string` has them ready:

```python
import string

print(string.punctuation)
print('(said)'.strip(string.punctuation))
print("Alice's".strip(string.punctuation))     # only the edges: the apostrophe inside stays
```

## Testing what a string contains

A family of methods answers yes-or-no questions about the *whole* string. You have seen `.isdigit()` and `.isalpha()`; here is the family:

| Method | `True` when every character is ... |
| --- | --- |
| `.isalpha()` | a letter |
| `.isdigit()` | a digit |
| `.isalnum()` | a letter or a digit |
| `.isspace()` | whitespace |
| `.isupper()` / `.islower()` | ... and the letters are all upper / all lower case |
| `.istitle()` | ... and every word starts with a capital |

Two things to keep in mind: they look at *all* characters (a single space makes `.isalpha()` `False`), and the empty string gives `False` for all of them.

```python
for item in ['Alice', 'alice', '1865', 'Alice1865', ' ', 'ALICE', '']:
    print(repr(item), item.isalpha(), item.isdigit(), item.isupper(), item.istitle())
```

(`repr()` shows the value with its quotes, so that the space and the empty string are visible.) Typical uses: `.isalpha()` to keep only "real" words when counting, `.isdigit()` to check user input before `int()`, `.isupper()` to spot shouting or acronyms, `.istitle()` to spot capitalised words, which are often names.

## Splitting and joining

`.split()` turns a string into a list. Without an argument it splits on any run of whitespace, including newlines, and drops empty pieces; with an argument it splits on exactly that separator, and keeps empty pieces. The difference matters:

```python
line = 'Emma, Jane Austen, 1815'

print(line.split())          # on whitespace
print(line.split(','))       # on the comma: note the leading spaces
print(line.split(', '))      # on comma and space

print('a  b'.split())        # two spaces: one separator
print('a  b'.split(' '))     # two spaces: an empty piece in between
```

`.splitlines()` splits on line breaks, which you know from the session on files. `.split()` has a second argument, the maximum number of splits, which is handy when only the first piece matters:

```python
entry = 'Emma: a novel by Jane Austen, first published in 1815'

print(entry.split(':'))
print(entry.split(':', 1)[0])     # at most one split, keep the first piece
```

The inverse of `.split()` is `.join()`, and its syntax surprises everyone: you call it on the **separator**, and pass the list. `', '.join(words)` glues the elements of `words` together with a comma and a space between them. All elements must be strings.

```python
words = ['The', 'Queen', 'of', 'Hearts']

print(' '.join(words))
print(', '.join(words))
print(''.join(words))
print('\n'.join(words))
```

`.join()` is how you get from a list of words back to a sentence, from a list of lines back to a text, and from a list of characters back to a string, which makes it the partner of every loop that collects characters or words. Two idioms:

```python
# reverse the words of a sentence
sentence = 'The Queen of Hearts she made some tarts'
print(' '.join(sentence.split()[::-1]))

# keep only the letters of a word
word = "Alice's"
letters = []
for char in word:
    if char.isalpha():
        letters.append(char)
print(''.join(letters))
```

## Tokenisation by hand

Everything in this session comes together in one task you will do constantly: turning a text into a clean list of words, or **tokens**. In the session on dictionaries you counted `alice,` and `alice` as different words; now you can do better. The recipe: lowercase, split on whitespace, strip the punctuation from the edges of every piece, drop what is empty.

```python
import string

def tokenize(text):
    """Return the lowercased words of text, without punctuation at their edges."""
    tokens = []
    for piece in text.lower().split():
        token = piece.strip(string.punctuation)
        if token:
            tokens.append(token)
    return tokens

print(tokenize('"Curiouser and curiouser!" cried Alice (she was much surprised).'))
```

```python
from collections import Counter

with open('../data/alice.txt', encoding='utf-8') as infile:
    alice = infile.read()

tokens = tokenize(alice)
counts = Counter(tokens)

print(len(tokens), len(counts))
print(counts['alice'])
print(counts.most_common(10))
```

Compare `counts['alice']` with the 223 you found by splitting alone: the difference is all the `alice,`, `alice.` and `alice!` that are now counted where they belong. This tokeniser is crude (it does not know that `don't` is one word and `well-known` perhaps two, and it does nothing about sentence boundaries), but it is honest and you understand every line of it. The next session, on regular expressions, gives you a more precise instrument for the same job.

## Comparing and sorting strings

`==` compares strings exactly, case included, so lowercase both sides when case should not matter. Strings can also be ordered with `<` and `>`, and therefore sorted, alphabetically, with one catch: capitals come before lowercase letters (all of `A`-`Z` sort before `a`), so `sorted()` puts `'Zebra'` before `'apple'`. To sort properly, lowercase the words first; the session on functional programming shows how to tell `sorted()` to do that for you.

```python
print('Alice' == 'alice', 'Alice'.lower() == 'alice'.lower())

words = ['banana', 'Zebra', 'cherry', 'apple']
print(sorted(words))
```

## Common mistakes

Four things that go wrong with strings every week.

```python
# Calling a method and forgetting to store the result: nothing happens
text = 'ALICE'
text.lower()
print(text)             # still ALICE
```

```python
# This cell produces an error on purpose: an index beyond the end of the string
word = 'Alice'
print(word[5])
```

```python
# This cell produces an error on purpose: .join() wants strings, not numbers
years = [1815, 1897, 1871]
print(', '.join(years))
```
