## 1. What is a notebook?

A Jupyter notebook is a document that mixes **text** and **code** that you can run. It is based on the idea of a [narrative](https://docs.jupyter.org/en/latest/use/use-cases/content-user.html): much like a scientist's lab notebook, which contains observations alongside interpretation, a notebook lets you tell a story in which the code and its results are part of the text. It is the standard working environment in data analysis, and it is what we use in every session of this programme.

The text you are reading now is written in **Markdown**, a simple way of formatting text. You can make text **bold** by wrapping it in double asterisks (`**bold**`), or make a heading by starting a line with one or more `#` characters. This [cheatsheet](https://github.com/adam-p/markdown-here/wiki/Markdown-Cheatsheet) has the rest.

### Cells

A notebook is divided into **cells**. This text sits in a cell of the type *Markdown*; the code further down sits in cells of the type *Code*. Click once on this text: the dropdown menu in the toolbar at the top now says "Markdown". Click on a code cell and it says "Code".

- To **run** a cell, click in it and press **Shift+Enter**: a Markdown cell is rendered, a code cell is executed and its result appears underneath. There are alternatives: **Ctrl+Enter** runs the cell without moving to the next one, **Alt+Enter** runs it and creates a new cell below, and the "play" button in the toolbar does the same as Shift+Enter.
- To **edit** a Markdown cell, double-click it. Try it on this cell, then run it again to render it.
- To **add** a cell, use the `+` button in the toolbar. Change its type with the dropdown menu.
- To **delete** a cell, select it and use the scissors button (or press `Esc` and then `D` twice).
- Explore the menus for the rest, and **save** your work regularly with the disk icon or **Ctrl+S** (**Cmd+S** on a Mac).

## 2. Hello, world!

Every programming language is traditionally introduced with a "Hello, world!" example. Run the following cell:

```python
# this will print some text
print("Hello, world!")
```

What happened here?

Python has a large set of **built-in functions**, and `print()` is one of them. When you use this function, `print()` writes its **argument** to the screen. "Argument" is a fancy word for the thing you put between the round brackets of a function. In this case, the argument is the **string** `"Hello, world!"`. A string is simply a sequence of characters, and the quotation marks tell Python where the string starts and ends.

Did you also notice the first line, which starts with a hash (`#`)? That is a **comment**. Comments document our code and explain what is happening; Python ignores them. We use them a lot in this course to make the code easy to follow.

## 3. Showing a value versus printing it

Run the following two cells and compare what appears underneath them:

```python
"Hello"
```

```python
print("Hello")
```

A notebook automatically **shows the value** of the last line of a code cell, with an `Out[...]` label. Python shows a string the way you would type it, with quotation marks. The function `print()`, on the other hand, **writes** the text to the screen, without quotation marks and without the `Out` label, and it works on any line of a cell, not just the last one. Two consequences that you will run into: only the *last* line of a cell is shown automatically (earlier lines are not, unless you print them), and if you don't want to see the value of the last line, you can end that line with a semicolon (`;`). Try both in the cell below:

```python
3 + 4     # not shown: not the last line
"Hello"   # shown: the last line

# Now add a semicolon after "Hello" and run the cell again.
```

For now, the difference between showing and printing is cosmetic; it becomes important later, when we talk about functions.

## 5. The kernel: order matters

Behind the notebook runs a Python process, called the **kernel**. It remembers everything you have run so far, *in the order in which you ran it*, which is not necessarily the order of the cells on the page. Run the two cells below, top to bottom:

```python
x = 5   # remember the number 5 under the name x (more about this in the next session)
```

```python
x * 2
```

Now choose **Kernel > Restart Kernel...** in the menu (this makes Python forget everything) and run *only* the second cell. You get a `NameError`: Python no longer knows what `x` is, because the cell that defines it has not been run since the restart. Run the first cell, then the second one, and all is well again.

Two habits that follow from this:

- If a cell complains that a name is not defined, look for the cell that defines it and run that one first.
- Before you hand in a notebook (and at the exam), choose **Kernel > Restart Kernel and Run All Cells**. If the notebook then runs from top to bottom without errors, it works for the person who reads it too.

If a cell keeps running forever (you see `[*]` next to it), the stop button in the toolbar or **Kernel > Interrupt** stops it.
