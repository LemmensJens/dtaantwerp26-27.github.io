A programming language is similar to a natural language: it has form and meaning, it even has idioms (programming clichés) and paraphrases (there are many ways to implement the same piece of code).

## Algorithms 

An algorithm is, informally, a finite series of steps that computes an output from a (possibly empty) input in a reasonable time. It is similar to the mathematical concept of a function, which also maps inputs to outputs.

Some examples of algorithms are:
- finding all palindromes in a text (input is a text, output is a list of palindromes),
- making a frequency dictionary or 'distribution' on the basis of a corpus (input is a list of texts, output is a dictionary of all the words in the corpus with their frequency),
- and so on.

## Example programming task
    
- Input: a word (e.g. 'Bootcamp')
- Output: a representation of the word as a pattern of consonants and vowels (in this case: 'CVVCCVCC')

How would you handle this?


## Some Ideas for a Solution

Semantics / algorithm:

- I have to know which letters are vowels and which ones are consonants (a **test**).
- I have to do this test for each letter of the word (a **loop**).
- I have to produce a 'C' or a 'V' in the output for each letter in the input (**input** and **output**).

Syntax / code (**don't worry about understanding the details yet**):

```python
def is_vowel(letter):
    """A test that checks whether a letter is a vowel or not."""
    if letter in ['a', 'e', 'i', 'o', 'u']:
        return True
    else:
        return False


def cv_convert(word):
    """Loop through the letters of a word, check whether each letter is a vowel,
    and add 'C' or 'V' to the result depending on the outcome of the test."""
    result = ''
    for letter in word:
        if is_vowel(letter):
            result += 'V'
        else:
            result += 'C'
    return result
```

```python
is_vowel('a')
```

```python
is_vowel('x')
```

```python
cv_convert('Hi there')
```

Notice:

- Syntax: punctuation (colons, brackets, quotes), indentation, keywords.
- Semantics: a test, a loop, ...
- Idiom: loop over a sequence and perform a test on each element of the sequence.

## Part 2: Variables

Variables are used in programs to give a name to something. In variable **assignment** we link a name to a value by writing *name = value*. The result is that the name is linked to that value: the name is **bound** to the value.

```python
publication_year = 1851
publication_year
```

If a name is used that has not been assigned a value, we get an error message, in this case a `NameError`, because this variable's name hasn't been defined yet:

```python
# This cell produces an error on purpose: the name has not been defined
sequel_year
```

A variable keeps its value unless we reassign it: we can overwrite the current value of a variable name by assigning it a new value.

```python
publication_year = 1851
```

```python
sequel_year = publication_year + 1
sequel_year
```

```python
author_age = publication_year - 1819   # Herman Melville was born in 1819
author_age
```

```python
author_age + 1
```

The single `=` is called the **assignment operator**. Variable assignment is not the same as **equality testing** (for which we use `==`, which we'll cover in greater depth later):

```python
x = 3 + 5
x
```

```python
x == 8
```

```python
x == 7
```

Finally, we should add that there exist a couple of **shortcut operators**, like `+=`, `-=`, `*=`, etc. These operators combine a basic arithmetic operation, like an addition, with a reassignment:

```python
year = 2020
year += 1
print(year)
```

This is completely equivalent to the following code block, but more concise and "Pythonesque":

```python
year = 2020
year = year + 1
print(year)
```

You can also link strings to variables:

```python
author = "Herman Melville"
print(author)
```

## Restrictions on the syntax of names

- All letters can be used, and the underscore (`_`); spaces, hyphens and other special characters cannot.
    - Strings (starting with `"` or `'`) can therefore not be used as names.
- Python is case-sensitive.
    - Uppercase and lowercase versions of a letter are different characters for Python: `Author` and `author` are two different names.
- Digits can be used, but not as the first character of a variable name.
    - Numbers can therefore not be used as names.
- Python **reserved words** or **keywords** (names that have a special meaning for Python) cannot be used.
    - Here's a list: **False, True, None, and, or, not, as, assert, break, continue, class, def, del, if, elif, else, except, finally, for, from, global, import, in, is, lambda, nonlocal, pass, raise, return, try, while, with, yield**
    - Luckily, they are colour-coded in most editors, including Jupyter.

## Programming style: variable names

- Use names that make clear what kind of *value* the variable holds: `age`, `birth_year`, `publication_date`, `name`, `location`, ...
- Do **not** name variables after their *type* (`string`, `list`, ...). This gets you into trouble later on.
- Use names that are not ambiguous: `first_name` and `family_name` rather than `name1` and `name2`.
- Words in a name are separated by underscores (`publication_year`, not `publicationYear`). This is the Python convention, and we use it throughout the course.
