## Part 1: Basic data types

### Numbers

For working with numbers in Python, two basic data types are available: **int** (for integer or "whole" numbers) and **float** (for floating-point numbers). You can always check for the type of something by passing it to the `type()` function:

```python
type(2)
```

```python
type(2.1)
```

Above, we apply this to a so-called **literal**, i.e. a hard-coded value in Python that doesn't get stored in a particular variable. The function `type()` can just as well be applied to a variable:

```python
a = 2.1
type(a)
```

A number of common **operators** are available to perform basic **arithmetic** in Python. These operators combine two variables or literals. Some simple examples:

```python
5 + 3 # addition
```

```python
5 - 2 # subtraction
```

```python
6 * 2 # multiplication
```

```python
2 ** 6 # power
```

```python
6 / 2 # division
```

The final example above demonstrates that the type of the result of an arithmetic operation can change, in the sense that, for instance, the result of the division of two integers will be a float. Also, notice what happens when we add a `float` to an `int`:

```python
type(6 + 2.0)
```

An arithmetic operation combines two numbers into a new number, but operations can of course be stacked to express more complex calculations:

```python
5 + 2 - 20 # of course, negative numbers exist too in Python!
```

In such cases, however, some operations will be carried out before others (as you will remember from your math courses in high school). This is called **operator precedence**:

```python
5 + 3 * 10 # predict the result?
```

If you'd like to take more control over the execution order of these operations, you can use round brackets:

```python
(5 + 3) * 10
```

#### Modulus and integer division

Two more operators are worth knowing. The **modulus** operator (`%`) gives the *remainder* of a division, and `//` gives the result of a division *rounded down* to a whole number. You won't need them often outside of programming exercises, but they come back regularly. Have a look at the examples below and describe their behaviour:

```python
print(13 // 6)
print(11 // 3)
print(11 / 3)   # compare with the ordinary division
```

Floats often have more decimals than you care about. The function `round()` rounds a number to the number of decimals you specify (or to a whole number if you don't specify any):

```python
print(round(11 / 3, 2))
print(round(11 / 3))
```

### Booleans

Another data type is the boolean (type `bool`). Booleans can only take one of two mutually exclusive values: **True** or **False**. Such binary values correspond to the zeros and ones (cf. "bits") that are common in computational architectures. It's a value that is "on" or "off":

```python
type(True)
```

```python
b = False
print(b)
```

Booleans are very important in programming, especially in the context of logical expressions (which are the topic of the session on `if` and `else`). You'll often want to check whether a certain condition holds "True": verifying the veracity of such a condition (e.g. is a password correct? is a password long enough?) is done using booleans. There are a number of **logical operators** that can help you with that. More on that in that session, but for now, we'll already mention the keywords `and`, `or` and `not`, which you can use to combine (or invert) logical values. (Remember: keywords are special "function words" in Python, that are part of the language itself and which shouldn't be used as e.g. variable names.)

You'll know from other contexts how two booleans can be combined into a new boolean by using these keywords. Some examples:

```python
print(True and True)
print(True and False)
print(False and True)
print(False and False)
```

Or:

```python
print(True or True)
print(True or False)
print(False or True)
print(False or False)
```

Inversion of a boolean can be done using the `not` keyword:

```python
print(not True)
print(not False)
```

Arguably the most common data type when dealing with digital text is the **string** (data type `str`). Strings are called that because Python considers them to be a "string", or an ordered sequence, of individual characters. String variables can be assigned from string literals that you can define in several ways:

```python
s = 'bootcamp' # single quotes
s = "bootcamp" # double quotes

type(s)
```

### Casting

Above we've discussed the four main primitive data types in Python and some of their core properties. Importantly, it is often possible to convert a variable to another data type. Converting a variable to another data type is very common in Python and this operation is called **casting**. Let us review a number of contexts in which casting is relevant.

First, we should emphasize that a number, for instance, could be represented both as a string (of a single character) and as an integer:

```python
a = 5     # int
b = '5'   # str
print(a)
print(b)
```

While these two 'numbers' might look very similar at the surface, the underlying data type greatly affects what you can do with such a variable. The first line below doesn't pose a problem, but the second line throws a **TypeError**, because subtraction is not something the string type supports:

```python
# This cell produces an error on purpose
print(a - 3)
print(b - 3)
```

Converting a string to an integer (or the other way round) changes what we can do with a variable.

A more practical example comes from the problem of **string formatting**. Often, you'll want to print some information to screen, where a part of the information is a predetermined template, and another part are numbers that have to be filled in dynamically at runtime (maybe because their value isn't known beforehand). For example:

```python
# This cell produces an error on purpose
age = 999
print('Saint Nicholas is ' + age + ' years old.')
```

Above, we try to combine a string literal, an integer (`age`), and another string literal. Our attempt fails, however, with a `TypeError`: this informs us that we cannot simply sum or concatenate a string and a number because their types don't match. It makes sense that Python is confused: do we intend the summation in the mathematical sense or in the string concatenation sense?

We can make that explicit by casting our `age` variable to a string before concatenating it:

```python
print('Saint Nicholas is ' + str(age) + ' years old.')
```

This works because the `str()` function will return a string variable containing the value `"999"`. Equivalent functions exist for casting to other data types. Below go a couple of common casting operations:

```python
v = int('999') # str to int
print(v)
print(type(v))
```

```python
v = float(999) # int to float
print(v)
print(type(v))
```

Depending on the exact goal that you have in mind, you might have to pay close attention to exact behaviour of these casting operations. The result of a casting operation can be unexpected at times:

```python
int(999.4) # implicit rounding to an integer
```

```python
# This cell produces an error on purpose
int('999.4') # not everything can be converted to an integer
```

```python
bool(0)
```

```python
bool('0')
```

### Printing several things at once

A quick reminder of how a code cell shows results. A notebook automatically displays the value of the **last line** of a cell (with an `Out` label); the values of earlier lines are not shown. `print()` shows something at any point in a cell, and is the only way to show several things from one cell. If you want to suppress the automatic display of the last line, end it with a semicolon:

```python
word = 'bootcamp'
word.upper()    # not shown: not the last line
word.title()    # shown
```

```python
print(word.upper())
print(word.title())
word.title();   # the semicolon suppresses the automatic display
```

Back to string formatting. Luckily, in many instances, you won't need to cast your variables to strings if you only care about printing them. Using commas, you can pass as many elements to `print()` as you wish:

```python
print(salutation, name, greeting)
print('Saint Nicholas is', age, 'years old.')
```

Python will automatically add a space in between the components before joining them into a single string. Adding our punctuation mark, however, is trickier now:

```python
print(salutation, name + '.', greeting)
```

A more powerful and very popular approach is the so-called **f-string** (or "formatted" string). You declare one by adding an `f` before the opening quotation mark of a string literal. Then you can insert other variables directly inside the string, between curly brackets. This is highly readable and you don't have to worry about casting:

```python
name = 'Mike'
s = f'Hi, my name is {name}.'
print(s)
print(f'Saint Nicholas is {age} years old.')
```

Inside the curly brackets you can put any expression, and after a colon you can say how a number should be displayed, for instance with two decimals:

```python
bad = 5
good = 7
print(f'I found {bad + good} mushrooms, {bad} of them poisonous.')
print(f'The proportion of poisonous mushrooms is {bad / (bad + good):.2f}.')
```

In older code you will often see the `.format()` method instead, which does the same with numbered or empty placeholders: `'I found {} mushrooms.'.format(bad + good)`. You should recognise it; in this course we use f-strings.

## Part 2: Lists and indexing

### Lists

**Lists** are a very common data type in Python to store *sequences* of items: a collection of elements that are ordered, so that each element can be uniquely identified by its position or **index** in the list. They can be manually defined using a syntax with square brackets and commas to separate the elements:

```python
l = ['a', 'b', 'c', 'd']
print(l)
```

In the example above, the elements are all unique and they all belong to the same type (`str`). Nevertheless, it is perfectly fine in Python to define lists in which the same element occurs multiple times; the elements don't even have to be of the same type, making this data type highly flexible:

```python
l = ['a', 'b', 'a', 'b']
print(l)

l = ['a', 1, False, 8.9]
print(l)
```

Lists can often contain other lists, in which case we speak of **nested lists**:

```python
l = [['a', 'b', 'c'], ['d', 'e', 'f'], ['g', 'h', 'i']]
print(l)
```

More on that later. You can also cast other data types to a list, using the `list()` function:

```python
l = list('bootcamp')
print(l)
```

Casting a string to a list gives you its *characters*. Much more often you want its *words*: the string method `.split()` cuts a string at the spaces and returns a list of the pieces. This is the simplest way to turn a text into words, and you will use it constantly:

```python
sentence = 'Call me Ishmael.'
words = sentence.split()
print(words)
```

Finally, a list can be empty. An empty list, `[]`, is often the starting point of a calculation: you create it first and add elements to it later (with `append()`, which we see below).

```python
result = []
print(result)
print(len(result))
```

To retrieve the **length** of a collection, use the `len()` function. After `print()`, this is probably the most often used function in Python. Applied to our list of characters and to our list of words:

```python
print(len(l))
print(len(words))
```

This function returns an integer expressing how many elements the list contains. It is also often applied to strings, which share a lot of characteristics with lists:

```python
s = 'bootcamp'
len(s)
```

One useful keyword in the context of lists is **`in`**: you can easily check whether some element is contained in a list or not. (In technical terms, we say that we **test for membership**.) The following lines yield booleans:

```python
l = ['a', 'b', 'c']
print('d' in l)
print('a' in l)
```

Additionally, you can apply the arithmetical operators `+` and `*` to concatenate or repeat lists. The behavior of these operators is completely analogous to what we saw with strings:

```python
print(l * 3) # repeat the list n times
print(l + ['d', 'e'])
```

A few built-in functions take a whole list as their argument. For a list of numbers, `sum()`, `min()` and `max()` do what their names say; `min()` and `max()` also work on a list of strings (alphabetically) and on a string itself (by character):

```python
numbers = [4, 8, 15, 16, 23, 42]
print(sum(numbers))
print(min(numbers))
print(max(numbers))
print(max(['pear', 'apple', 'quince']))
```

Finally, lists have many **methods**, functions that you call on the list itself with the dot syntax, as we saw for strings. Below, we demonstrate a number of them. Can you describe their behaviour?

```python
l = list('bootcamp')
print(l.count('o'))
print(l.count('k'))
print(l.index('t'))   # the position of the first 't'
```

```python
l.pop()
print(l)
```

```python
l.append('p')
print(l)
```

```python
l.reverse()
print(l)
l.reverse()
```

```python
l.remove('o')
print(l)
```

```python
l.sort()
print(l)
```

Note that `sort()`, like `reverse()`, `append()` and `remove()`, changes the list **in place** and returns nothing: the cell above had to `print(l)` on a separate line to show the result. If you want a sorted *copy* and leave the original list alone, use the general function `sorted()` instead, which works on any sequence (strings too) and returns a new list:

```python
l = list('bootcamp')
print(sorted(l))
print(l)                       # unchanged
print(sorted(l, reverse=True)) # descending order
```

### Indexing

Retrieving specific elements from sequences like lists and strings is an important Python skill. This is called **indexing**. Remember that any item in a sequence is uniquely defined by its position or **index**. If we would like to retrieve a specific item from a list, we therefore use that index, with square brackets:

```python
l = ['a', 'b', 'c', 'd', 'e']

print(l[0])
print(l[1])
print(l[2])
print(l[3])
print(l[4])
```

A number of crucial observations must be made here. First of all, you'll have to get used to the somewhat weird fact that Python is a **zero-indexed** language, which means that it always starts counting from zero. With indexing too, the index of the very first element in a list is zero because of that. All the other indices will always, intuitively speaking, seem to be "off" by one, i.e. if you want to access the fifth character, you'll need index four. Also convince yourself of the following fact: the index of the final element, consequently, will always be its length minus one:

```python
l[len(l) - 1]
```

Luckily, Python also has a mechanism to index sequences "from the rear", using negative indices:

```python
print(l[-1])
print(l[-2])
print(l[-3])
print(l[-4])
print(l[-5])
```

Unfortunately, because -0 doesn't exist, here we *do* start counting from -1. Because of that, you should convince yourself of the fact that you can retrieve the first element from a sequence by negating its length and using that as an index:

```python
l[-len(l)]
```

Because strings and lists are both sequences in Python, indexing works the same way for both:

```python
l = 'bootcamp'
print(l[0])
print(l[1])
print(l[2])
print('...')
print(l[-3])
print(l[-2])
print(l[-1])
```

Using a single index will yield a single element. If you would like to get a subsequence or **slice** from your sequence, however, that is also possible. For that, we use the following syntax:

```python
print(l[0:3])
print(l[3:6])
print(l[5:6])
```

With the colon (`:`), we can first specify a start index and, next, an end index. The confusing part is that the indexing mechanism in Python is **upper bound exclusive**: it will take everything up to, but *excluding* the end index. This too is something you will have to get used to.

A very nice feature of Python is that it allows to leave out the start or end index. In these cases, the rest of the sequence will be returned all the way up the beginning or end of the sequence:

```python
print(l[4:])
print(l[:4])
```

This will also work for negative indices; to get the final two characters of a string:

```python
l[-2:]
```

**Test your understanding**: Consider the sequence `s = [1, 2, 3, 4, 5]`:
- Will the expression `s[:i] + s[i:]` be equal to the original sequence?
- What will be the length of the slice `s[2:4]`?

Apart from the start and end index it is possible to specify a third number when slicing, the so-called **step size**. With this parameter, you specify how many elements to skip in between:

```python
print(l[::1]) # nothing happens
print(l[::2]) # only the characters at even indices (0, 2, 4, ...)
```

The following (admittedly: somewhat complex) feature is rarely used, but it does have one very nice application. Can you explain what the following operation does?

```python
l[::-1]
```

There are of course other ways to achieve this (you already know one alternative!) but this notation is surprisingly concise and practical.

Knowing about the start index, end index and step size is also useful for initializing so-called **ranges** of numbers. Below go a couple of examples that show how you can use the function `range()` to generate arbitrary sequences of integers:

```python
list(range(10))
```

```python
list(range(5,10))
```

```python
list(range(5,10,2))
```

### A word on mutability

An important concept when talking about collections is **mutability** or the freedom that you have in changing the values in individual positions in sequences. Generally speaking, Python is very relaxed in redefining entire variables. You can perfectly overwrite the value of a previously defined variable and you can even freely change its type -- no problem there:

```python
n = 'abra'
n = 'cadabra'
n = 3
```

Redefining an *entire* string variable is not a problem. The following, however, does not work with strings:

```python
# This cell produces an error on purpose
l = 'abracadabra'
l[0] = 'e'
print(l)
```

We are trying to redefine the value of an element in a specific position in the string, but Python refuses to do that. The surprising thing is that this does work for lists:

```python
l = list('abracadabra')
l[0] = 'e'
print(l)
```

This difference relates to mutability: lists are said to be **mutable** in Python, whereas strings are not. For the sake of completeness, we should also mention the existence of **tuples**. Tuples look a lot like lists and they support the same kind of indexing as the other sequences which we already covered. You can define them using round brackets:

```python
l = tuple('abracadabra')
print(type(l))
print(l)
```

The main difference between lists and tuples is that tuples are immutable: once created, they cannot be changed. They are used for small, fixed groups of values (a title and a year, for instance: `('Moby-Dick', 1851)`). You will see them appear now and then; it might take a while before you need to create one yourself.
