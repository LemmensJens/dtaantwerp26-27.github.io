## Dictionaries and counting

Lists store things in a fixed order and you reach an element by its position. Very often, though, what you want is to reach a value by a **name**: the year of a novel by its title, the number of times a word occurs by the word itself. That is what a **dictionary** does, and it is the data structure you will use most in this programme after strings and lists. This session introduces dictionaries and their small cousin, the set, and then puts them to work on the most classic task in text analysis: counting how often things occur.

After this session you can:

- create a dictionary, look up, add, change and remove entries, and loop over it;
- use a set to get the unique elements of a collection;
- build a frequency distribution of characters or words with a dictionary, and read off the most frequent ones;
- catch an error with `try` and `except` instead of letting it stop your program;

## Dictionaries

A dictionary stores **key-value pairs**: every key maps to a value, and you use the key to find the value, the way you use a word to find its definition in a real dictionary. Dictionaries are written with curly brackets, each pair as `key: value`, pairs separated by commas. Here is one that describes a medieval manuscript:

```python
manuscript = {'material': 'parchment',
              'leaves': 193,
              'language': 'Latin',
              'damaged': True}
print(manuscript)
print(type(manuscript))
```

The keys here are strings, the values are a string, an integer, another string and a boolean: the values can be of any type, and they need not all be the same. You look up a value with the key between square brackets, the same brackets you use for indexing a list, except that the "index" is now a key:

```python
print(manuscript['material'])
print(manuscript['leaves'])
```

A key that is not in the dictionary gives a `KeyError`. Read the message: it tells you which key was missing.

```python
# This cell produces an error on purpose: there is no such key
print(manuscript['weight'])
```

### Changing, adding and removing entries

Assigning to a key with the same bracket syntax **changes** the value if the key exists, and **adds** a new pair if it does not. This is how dictionaries grow:

```python
manuscript['leaves'] = 194          # change an existing value
manuscript['weight'] = 0.82         # add a new key-value pair
print(manuscript)
```

`del` removes a pair, and `len()` tells you how many pairs there are:

```python
del manuscript['damaged']
print(manuscript)
print(len(manuscript))
```

Two rules about keys. Every key is **unique**: assigning to a key that exists overwrites the old value rather than adding a second pair. And keys must be of an unchangeable type: strings, numbers and tuples are fine, lists are not (you will see the error at the end of this session). Values have no such restriction.

### Is it in there?

The keyword `in` tests whether a **key** is present. It looks at the keys only, not at the values:

```python
print('leaves' in manuscript)
print('parchment' in manuscript)
print('parchment' in manuscript.values())
```

Testing before you look up is the standard way to avoid a `KeyError`:

```python
key = 'weight'

if key in manuscript:
    print(manuscript[key])
else:
    print(f'no information about {key}')
```

### Looping over a dictionary

A `for` loop over a dictionary walks through its **keys**; you get the value by looking the key up:

```python
for key in manuscript:
    print(key, manuscript[key])
```

More convenient is `.items()`, which hands you the key and the value together as a pair. You catch the pair with two loop variables, exactly as with `enumerate()` and `zip()` in the session on loops:

```python
for key, value in manuscript.items():
    print(f'{key}: {value}')
```

`.keys()` and `.values()` give you just the keys or just the values, for instance to loop over the values alone, or to sort the keys: `sorted(manuscript)` returns the keys in alphabetical order.

```python
print(manuscript.keys())
print(manuscript.values())
print(sorted(manuscript))
```

### Building a dictionary in a loop

You will rarely type a dictionary out by hand. The usual pattern is: start with an empty dictionary, `{}`, and add pairs in a loop. Here we map every word in a list to its length:

```python
words = ['Alice', 'was', 'beginning', 'to', 'get', 'very', 'tired']

lengths = {}
for word in words:
    lengths[word] = len(word)

print(lengths)
print(lengths['beginning'])
```

### Values can be anything

A value can itself be a list, a tuple or another dictionary, and that is where dictionaries become really useful. A dictionary with **list values** stores several things per key:

```python
characters = {'Alice': ['Alice'],
              'Emma': ['Emma', 'Mr Knightley', 'Harriet'],
              'Dracula': ['Dracula', 'Jonathan', 'Mina', 'Van Helsing']}

print(characters['Dracula'])
print(len(characters['Dracula']))
characters['Alice'].append('the White Rabbit')    # the value is a list, so list methods work on it
print(characters['Alice'])
```

A dictionary with **tuple values** keeps a fixed group of facts per key, which you unpack when you look them up:

```python
novels = {'Emma': ('Jane Austen', 1815),
          'Dracula': ('Bram Stoker', 1897),
          'Middlemarch': ('George Eliot', 1871)}

author, year = novels['Dracula']
print(f'Dracula was written by {author} and published in {year}.')
```

And a **list of dictionaries** is the natural way to represent a table: one dictionary per row, with the column names as keys. This is what you get when you read a spreadsheet or a CSV file row by row, and it is the shape of most data you will download in this programme.

```python
table = [{'title': 'Emma', 'author': 'Jane Austen', 'year': 1815},
         {'title': 'Dracula', 'author': 'Bram Stoker', 'year': 1897},
         {'title': 'Middlemarch', 'author': 'George Eliot', 'year': 1871}]

for row in table:
    print(row['title'], row['year'])
```

Such a row is easy to build from a header and a line of values with `zip()`, which pairs them up, and `dict()`, which turns the pairs into a dictionary:

```python
header = ['title', 'author', 'year']
values = ['Persuasion', 'Jane Austen', 1817]

row = dict(zip(header, values))
print(row)
```

## Sets

A **set** is a collection of **unique** elements without any order: the opposite of a list, which keeps duplicates and order. Sets are written with curly brackets too, but without the colons. Duplicates disappear the moment the set is created:

```python
letters = {'a', 'b', 'a', 'c', 'b'}
print(letters)
```

The typical use is to turn a list or a string into a set with `set()`, to find out **which different elements** it contains, or **how many**:

```python
words = ['the', 'cat', 'sat', 'on', 'the', 'mat', 'the', 'end']

print(set(words))
print(len(words), len(set(words)))
```

Number of words versus number of *different* words (in linguistics: tokens versus types) is a first, crude measure of how varied a text is. A set has no order, so it cannot be indexed; `sorted()` gives you its elements as an ordered list, and `in` tests membership as with lists. Sets can also be compared (`.intersection()`, `.difference()`, `.union()`): try `help(set)` if you need that.

```python
print(sorted(set(words)))
print('cat' in set(words))
```

## Counting things: the evergreen

Now the classic. Every text-analysis course begins with the same exercise: count how often each character occurs in a text, using a dictionary. Our text is the opening of Virgil's *Aeneid*:

```python
aeneid = """Arma virumque cano, Troiae qui primus ab oris
Italiam fato profugus Lavinaque venit
litora—multum ille et terris iactatus et alto
vi superum, saevae memorem Iunonis ob iram,
5 multa quoque et bello passus, dum conderet urbem
inferretque deos Latio; genus unde Latinum
Albanique patres atque altae moenia Romae.
Musa, mihi causas memora, quo numine laeso
quidve dolens regina deum tot volvere casus
10 insignem pietate virum, tot adire labores
impulerit. tantaene animis caelestibus irae?"""

print(aeneid)
```

The idea: the dictionary maps every character to its count. For each character in the text, if it is already in the dictionary, add one to its count; if not, it is the first time we see it, so start its count at 1.

```python
frequencies = {}

for char in aeneid:
    if char in frequencies:
        frequencies[char] += 1
    else:
        frequencies[char] = 1

print(frequencies)
```

Look at the result: the space is the most frequent "character", upper and lower case are counted separately, and the punctuation and the line numbers are in there too. Whether that is what you want depends on your question, and the fix is always the same: a test inside the loop that decides what to count, and a `.lower()` where needed.

### Reading off the results

A dictionary of counts is only useful if you can find the interesting entries. Looping over `.items()` with a test does most of the work: for instance, all words that occur more than once.

```python
word_counts = {}
for word in aeneid.lower().split():
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

for word, count in word_counts.items():
    if count > 1:
        print(word, count)
```

To find the *single* most frequent entry, keep track of the best one seen so far, the way you found the longest line in the exercises on files:

```python
most_frequent = ''
highest_count = 0

for word, count in word_counts.items():
    if count > highest_count:
        most_frequent = word
        highest_count = count

print(most_frequent, highest_count)
```

And to see the whole distribution in order, sort it: `sorted(word_counts)` sorts the keys alphabetically. Sorting by *count* is a bit more work in plain Python; you will see the easy way at the end of this session.

```python
for word in sorted(word_counts):
    print(word, word_counts[word])
```

## Errors you can catch: `try` and `except`

Look once more at the counting loop. The `if`/`else` is there because we cannot add 1 to a count that does not exist yet: the first time a character comes up, `frequencies[char] += 1` would look up a missing key and stop the program with a `KeyError`:

```python
# This cell produces an error on purpose: the key does not exist yet
frequencies = {}

for char in aeneid:
    frequencies[char] += 1
```

In the session on `if` and `else` you saw that an error like this is called an **exception**, that it is *raised*, and that you can raise one yourself. Here is the other side of that story: Python lets you **catch** an exception and decide what to do instead of stopping. The keywords are `try` and `except`, and they look a lot like `if` and `else`: Python *tries* to execute the first block; if an exception is raised in it, it jumps to the `except` block instead of stopping.

```python
frequencies = {}

for char in aeneid:
    try:
        frequencies[char] += 1
    except KeyError:
        frequencies[char] = 1

print(frequencies)
```

The result is the same as with `if`/`else`. Note that the `except` names the exception it is prepared to handle: `except KeyError:`. Always do that. A bare `except:` catches *every* error, including the ones you did not anticipate (a typo in a variable name, say), and then silently does the wrong thing instead of telling you. Name the exception you expect, and let everything else stop your program the way it should.

A more typical use of `try`/`except` is input that may or may not be usable. `int('seven')` raises a `ValueError`. Combined with the "keep asking" loop from the session on files, that gives a robust way to ask for a number:

```python
while True:
    answer = input('How many pages? ')
    try:
        pages = int(answer)
        break
    except ValueError:
        print('Please enter a whole number.')

print(f'{pages} pages')
```

Another everyday example is a list of values that *should* be numbers but contains a few gaps, as real data always does:

```python
raw_values = ['12', '7', 'N/A', '30', '', '4']

values = []
for raw in raw_values:
    try:
        values.append(int(raw))
    except ValueError:
        print(f'skipping "{raw}"')

print(values)
```

Remember the rule from the session on `if` and `else`: an error message is information. `try`/`except` is for cases where you *know* which error can occur and what to do about it, not for making error messages go away.

## `Counter`

You have written the counting loop several times now, which was the point: you should understand exactly what it does. Python's standard library has a ready-made version of it, `Counter`, in the module `collections`. It is imported a little differently from what you saw with `random` and `os`:

```python
from collections import Counter
```

`from collections import Counter` takes *one thing* out of the module and makes it available under its own name, so that you write `Counter` rather than `collections.Counter`. This form is common for modules from which you need a single function or type.

A `Counter` is a dictionary that is specialised in counting. Give it a sequence and it counts the elements:

```python
counts = Counter(aeneid.lower().split())
print(counts)
```

Everything you know about dictionaries applies: `counts['et']`, `.items()`, `in`, `len()`. Two things are new. First, the method `.most_common()`, which returns the entries sorted by count, most frequent first, as a list of (element, count) tuples. With a number as argument, it returns only the top of the list:

```python
print(counts.most_common(5))
```

Second, a `Counter` does not raise a `KeyError` for a missing key: it answers 0, which is the right answer for "how many times does this occur". That is also why the `if`/`else` is not needed when you count with it by hand:

```python
print(counts['dolens'])
print(counts['hobbit'])

char_counts = Counter()
for char in aeneid.lower():
    if char.isalpha():
        char_counts[char] += 1

print(char_counts.most_common(3))
```

`Counter` will be your standard tool for frequencies from now on. The whole *Alice* word frequency list is one line, and the ten most frequent words are one more:

```python
alice_counts = Counter(alice.lower().split())

print(alice_counts.most_common(10))
```

```python
# This cell produces an error on purpose: looking up a key that is not there
years = {'Emma': 1815, 'Dracula': 1897}
print(years['Frankenstein'])
```

```python
# This cell produces an error on purpose: a list cannot be a key ("unhashable" means "cannot be used as a key")
years = {}
years[['Emma', 'Persuasion']] = 'Jane Austen'
```
