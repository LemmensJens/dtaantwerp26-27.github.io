## Comprehensions and functions as arguments

By now you have written the same four lines dozens of times: make an empty list, loop over something, test each element, append. Python has a shorter way of writing exactly that pattern, the **list comprehension**, and you will see it in every notebook, every tutorial and every code base from here on. This session teaches you to read and write comprehensions, and then looks at a second idea that sits behind a lot of compact Python: passing a **function as an argument** to another function, which is how `sorted()` sorts by whatever you like, and how a dictionary finally gets sorted by its values.

After this session you can:

- write a list comprehension that transforms, filters, or both, and read one written by someone else;
- write a dictionary comprehension and a set comprehension;
- sort a list of anything with `sorted(..., key=...)`, including a dictionary by its values;
- write a small `lambda` function where a named function would be overkill;
- recognise `map()` and `filter()` when you meet them, and know that a comprehension does the same.

## From loop to comprehension

Here is the idiom, once more: the lengths of a list of words.

````python
words = 'the quick brown fox jumps over the lazy dog'.split()

lengths = []
for word in words:
    lengths.append(len(word))

print(lengths)
````

And here is the same thing as a **list comprehension**:

````python
lengths = [len(word) for word in words]

print(lengths)
````

Read it from the `for`: *for every `word` in `words`, take `len(word)`, and collect the results in a list*. The square brackets say "this builds a list"; inside them come an **expression** (`len(word)`) and a **loop** (`for word in words`). The loop variable is yours to name, as in a normal loop, and the expression can be anything that uses it:

````python
print([word.upper() for word in words])
print([word[0] for word in words])
print([f'{word}: {len(word)}' for word in words])
print([len(word) > 4 for word in words])
````

### Filtering

The loop idiom often has an `if` in it: keep only the elements that pass a test. In a comprehension, the `if` goes at the end:

````python
long_words = []
for word in words:
    if len(word) > 4:
        long_words.append(word)
print(long_words)

long_words = [word for word in words if len(word) > 4]
print(long_words)
````

Here the expression is just `word`: nothing is transformed, elements are only selected. Transforming and filtering combine, with the expression at the front and the test at the back:

````python
print([word.upper() for word in words if len(word) > 4])
print([len(word) for word in words if word.startswith('t')])
````

So a comprehension has up to three parts, always in this order:

```python
[expression  for element in sequence  if condition]
```

*What to collect*, *from where*, *under which condition*. The sequence can be anything you can loop over: a list, a string, a `range()`, a file, a dictionary, the result of `re.findall()`.

````python
print([n * n for n in range(1, 11)])
print([n for n in range(1, 31) if n % 3 == 0])
print([char for char in 'Wonderland' if char.lower() in 'aeiou'])
````

### Comprehensions and text

Comprehensions are made for the small cleaning steps that text needs. Reading the non-empty lines of a file, stripped, is one line; so is lowercasing tokens, or keeping the ones that consist of letters.

````python
import re
import string
from collections import Counter

with open('../data/alice.txt', encoding='utf-8') as infile:
    lines = [line.strip() for line in infile if line.strip()]

print(len(lines))
print(lines[:3])
````

````python
text = ' '.join(lines)

tokens = [token.lower() for token in re.findall(r'[A-Za-z]+', text)]
print(len(tokens))

long_tokens = [token for token in tokens if len(token) > 8]
print(Counter(long_tokens).most_common(5))
````

Compare the second cell with the `tokenize()` function you wrote in the session on strings: same idea, fewer lines. Both are fine; a comprehension wins when the whole step fits on one readable line, a loop wins when there is more to do per element.

### Choosing between two values

Sometimes you do not want to *drop* elements that fail a test but *replace* them. That is an `if`/`else` **inside the expression**, at the front, which is a different construction from the filtering `if` at the back:

````python
print(['long' if len(word) > 4 else 'short' for word in words])
print([word if len(word) > 3 else '_' for word in words])
````

Front `if`/`else`: every element produces a value, one of two. Back `if`: some elements produce nothing. You can even combine them, but at that point a loop is usually clearer.

### Nested loops

A comprehension can contain two `for` parts, which work like nested loops, outer one first. Use this sparingly: it is compact, but hard to read. The letters of all capitalised words, for instance:

````python
titles = ['Emma', 'Middlemarch', 'Dracula']

letters = [letter for title in titles for letter in title if letter.lower() in 'aeiou']
print(letters)
````

## Dictionary and set comprehensions

The same construction with curly brackets builds a **dictionary** (`key: value` before the `for`) or a **set** (a single expression). A dictionary comprehension is the compact form of the "build a dictionary in a loop" pattern from the session on dictionaries:

````python
word_lengths = {word: len(word) for word in words}
print(word_lengths)

first_letters = {word[0] for word in words}
print(first_letters)
````

````python
years = {'Emma': 1815, 'Dracula': 1897, 'Persuasion': 1817, 'Middlemarch': 1871}

nineteenth = {title: year for title, year in years.items() if year >= 1800}
by_year = {year: title for title, year in years.items()}      # the reverse dictionary
print(nineteenth)
print(by_year)
````

## Sorting by what you want: `key=`

`sorted()` sorts numbers by size and strings alphabetically, and you have met its limits twice: capitals sort before lower-case letters, and a dictionary cannot be sorted by its values. The remedy is the same in both cases. `sorted()` takes an optional argument `key=`, a **function**: before comparing two elements, `sorted()` calls that function on each of them and compares the *results*. So `key=len` sorts by length, and `key=str.lower` sorts by the lowercased version:

````python
titles = ['Middlemarch', 'emma', 'Dracula', 'Kidnapped', 'Persuasion']

print(sorted(titles))
print(sorted(titles, key=str.lower))
print(sorted(titles, key=len))
print(sorted(titles, key=len, reverse=True))
````

Note what is passed: `len`, not `len()`. You hand over the function itself, without calling it; `sorted()` calls it, once per element. A function is a value like any other in Python, and it can be stored in a variable, put in a list, or passed to another function. This idea is called **functional programming**, and `key=` is the place where you will use it most.

### A dictionary sorted by its values

The pairs of a dictionary come from `.items()` as tuples, `(key, value)`. To sort them by the value, the key function must return the second element of the pair. There is no built-in function for that, so we write a tiny one:

````python
counts = Counter(tokens)

def get_count(pair):
    return pair[1]

for word, count in sorted(counts.items(), key=get_count, reverse=True)[:5]:
    print(word, count)
````

That is exactly what `Counter.most_common()` does, and it works for *any* dictionary, not just a `Counter`.

### `lambda`: a function without a name

`get_count()` is three lines for something that is used once, in one place. Python has a shorthand for such throw-away functions: **`lambda`**, followed by the parameter(s), a colon and *one expression*, whose value is returned. `lambda pair: pair[1]` is `get_count` without the name:

````python
top = sorted(counts.items(), key=lambda pair: pair[1], reverse=True)[:5]
print(top)

print(sorted(titles, key=lambda title: title[-1]))        # by last letter
print(sorted(years.items(), key=lambda pair: pair[1]))    # novels by year
````

`lambda` cannot contain statements (no loops, no `if` blocks, no `return`), only one expression, which is why it suits `key=` and little else. When the function needs more than one line, or a name that explains it, write a normal `def`.

Two related built-ins take `key=` too: `max()` and `min()`. The most frequent word without sorting anything:

````python
print(max(counts.items(), key=lambda pair: pair[1]))    # the (word, count) pair with the highest count
print(min(titles, key=len))
````

## `map()` and `filter()`

Two older functions apply a function to every element (`map`) or keep the elements for which a function returns `True` (`filter`). They are what comprehensions replaced, and you should recognise them in code you read. They return a lazy object, not a list, so you wrap them in `list()` to see the result:

````python
def is_long(word):
    return len(word) > 4

print(list(map(len, words)))
print(list(filter(is_long, words)))
print(list(map(lambda word: word.upper(), words)))
````

````python
# the same, as comprehensions: this is the form to write
print([len(word) for word in words])
print([word for word in words if is_long(word)])
print([word.upper() for word in words])
````

Write the comprehension; it says the same thing and reads better. `map()` and `filter()` survive in older code and in a few idioms (`list(map(int, line.split()))` to turn a line of numbers into integers is one you will meet).

## When to use a comprehension, and when not

A comprehension is good when the whole step fits on one line that you can read aloud: *the lengths of the words*, *the lines that are not empty*, *the tokens longer than eight letters*. It is bad when you find yourself nesting two `for`s and two `if`s, or when the expression itself needs a comment. In that case write the loop: the extra lines cost nothing and the next reader, probably you, will be grateful. Compact is not the same as clear.

## Common mistakes

Three things that go wrong with comprehensions and keys.

````python
# Round brackets instead of square ones: not a list but a generator object
lengths = (len(word) for word in words)
print(lengths)
print(list(lengths))
````

````python
# This cell produces an error on purpose: key= wants a function, not the result of calling one
print(sorted(titles, key=len(titles)))
````

````python
# This cell produces an error on purpose: the filtering if goes at the end, the if/else at the front
print([word for word in words if len(word) > 4 else 'short'])
````
