## Regular expressions

With string methods you can find, count and replace a **fixed** piece of text: `text.count('Alice')`, `text.replace('Queen', 'King')`. But many questions are about a *kind* of text rather than a fixed one: every four-digit number, every capitalised word, every word ending in *-ing*, everything between quotation marks. For those you need a way to describe a **pattern**, and that is what **regular expressions** (regexes) are: a small language for describing patterns in text, which Python's `re` module can search for, extract, replace and split on.

Regular expressions have a reputation for being unreadable, and long ones are. The ones you will write in this programme are short, and the trick is to build them up one piece at a time and test each step. After this session you can:

- write a pattern with character classes, quantifiers and anchors, and read one written by someone else;
- find all matches of a pattern with `re.findall()`, and the first one with `re.search()`;
- clean text with `re.sub()` and split it with `re.split()`;
- pull a part out of a match with a group;
- write every pattern as a raw string, and know why.

## The `re` module and `re.findall()`

Everything lives in the module `re`. Its most useful function is `re.findall(pattern, text)`, which returns a list of all pieces of `text` that match `pattern`. With a pattern that is just ordinary characters, it behaves like a search for a fixed string:

```python
import re

text = 'Alice was beginning to get very tired of sitting by her sister on the bank.'

print(re.findall('sister', text))
print(re.findall('ing', text))
print(re.findall('Rabbit', text))
```

One convention before we go on: from now on every pattern is written as a **raw string**, `r'...'`, which you met in the session on strings. Patterns are full of backslashes, and inside a normal string Python would try to interpret them as escape sequences (`'\b'` is a backspace character to Python, but a word boundary to `re`). A raw string hands the backslashes to `re` untouched. The rule of this course: **every pattern is a raw string**, even when it contains no backslash yet.

```python
print(re.findall(r'ing', text))
```

## Building patterns

The power comes from **special characters** that stand for a set of possibilities rather than for themselves. We introduce them a few at a time, and test each on the same sentence.

### Character classes: one character out of a set

Square brackets match **one** character out of the set inside them. A dash gives a range, and a `^` right after the opening bracket *negates* the set.

```python
print(re.findall(r'[aeiou]', text))          # any vowel
print(re.findall(r'b[aey]', text))           # b followed by a, e or y
print(re.findall(r'[A-Z]', text))            # any capital letter
print(re.findall(r'[^a-z ]', text))          # anything that is not a lower-case letter or a space
```

Some sets are so common that they have a shorthand:

| Shorthand | Matches | Same as |
| --- | --- | --- |
| `\d` | a digit | `[0-9]` |
| `\w` | a "word character": a letter, a digit or `_` | `[a-zA-Z0-9_]` (plus accented letters) |
| `\s` | a whitespace character: space, tab, newline | `[ \t\n\r\f\v]` |
| `.` | any character except a newline | |

Capitalised, `\D`, `\W` and `\S` mean the opposite: *not* a digit, *not* a word character, *not* whitespace.

```python
dates = 'Emma appeared in 1815, Dracula in 1897 and Middlemarch in 1871-72.'

print(re.findall(r'\d', dates))
print(re.findall(r'\W', dates))
print(re.findall(r'a.', dates))              # a followed by any one character
```

### Quantifiers: how many times

A class or character matches exactly one character. A **quantifier** after it says how many times it may repeat:

| Quantifier | Meaning |
| --- | --- |
| `+` | one or more |
| `*` | zero or more |
| `?` | zero or one (optional) |
| `{4}` | exactly four |
| `{2,4}` | two to four |
| `{2,}` | two or more |

```python
print(re.findall(r'\d+', dates))            # one or more digits: whole numbers
print(re.findall(r'\d{4}', dates))          # exactly four digits: years
print(re.findall(r'\w+', dates))            # one or more word characters: words
print(re.findall(r'[A-Z]\w*', dates))       # a capital followed by zero or more word characters
print(re.findall(r'colou?r', 'colour, color, colr'))     # the u is optional
```

`\w+` is the simplest regex tokeniser: every run of word characters. Compare it with the `tokenize()` you wrote by hand in the previous session.

### Anchors: where in the text

Anchors match a *position*, not a character. `^` is the start of the text, `$` its end, and `\b` a **word boundary**: the position between a word character and a non-word character (or the start or end of the text). `\b` is the one you will use constantly, because it lets you say "as a whole word":

```python
sentence = 'The cat sat on the catalogue, concatenating cats.'

print(re.findall(r'cat', sentence))          # every occurrence, also inside other words
print(re.findall(r'\bcat\b', sentence))      # cat as a whole word
print(re.findall(r'\bcat\w*', sentence))     # words that start with cat
print(re.findall(r'\w+ing\b', sentence))     # words that end in ing
print(re.findall(r'^The', sentence), re.findall(r'cats\.$', sentence))
```

### Alternatives and groups

A vertical bar `|` means "or", and round brackets **group** a part of the pattern, for instance to apply a quantifier or an alternative to more than one character:

```python
print(re.findall(r'cat|dog', 'The cat and the dog and the catalogue.'))
print(re.findall(r'\b(?:cat|dog)s?\b', 'The cat, the dogs and the catalogue.'))
```

The `(?:...)` is a group that only groups. Plain brackets, `(...)`, also **capture**: they make `re.findall()` return only the part inside the brackets instead of the whole match, which is useful when you want to extract a piece from a larger context. Here we want the word that follows *the*:

```python
print(re.findall(r'\bthe (\w+)', 'The cat, the dogs and the catalogue.'))
```

### Special characters as themselves

`. ^ $ * + ? { } [ ] \ | ( )` all have a meaning in a pattern. To match one of them literally, put a backslash in front of it: `\.` is a full stop, `\?` a question mark, `\(` an opening bracket. Forgetting this is the most common regex bug: `re.findall(r'.', text)` does not find full stops, it finds everything.

```python
print(len(re.findall(r'.', text)))           # any character: as many matches as the text has characters
print(re.findall(r'\.', text))               # a literal full stop
print(re.findall(r'\(\d{4}\)', 'Emma (1815) and Dracula (1897)'))
```

### Ignoring case

The optional third argument of the `re` functions is a **flag**; `re.IGNORECASE` makes the pattern case insensitive, which is usually simpler than writing `[Aa]lice`.

```python
print(re.findall(r'alice', 'Alice asked. ALICE shouted. alice whispered.'))
print(re.findall(r'alice', 'Alice asked. ALICE shouted. alice whispered.', re.IGNORECASE))
```

## The other functions of `re`

`re.findall()` gives you the matching strings. The other functions give you more information, change the text, or split it.

### `re.search()`: the first match, with its position

`re.search(pattern, text)` returns the **first** match as a **match object**, or `None` if there is no match. The match object knows what was matched (`.group()`) and where (`.start()`, `.end()`, `.span()`). Because it is `None` when nothing matches, test it before you use it:

```python
match = re.search(r'\d{4}', dates)

print(match)
print(match.group())
print(match.start(), match.end(), match.span())
```

```python
match = re.search(r'\d{4}', text)

if match:
    print('found', match.group())
else:
    print('no year in this text')
```

If the pattern has capturing groups, `.group(1)` gives you the first group, `.group(2)` the second, and `.group()` (or `.group(0)`) the whole match. This is how you take a match apart:

```python
match = re.search(r'(\w+) appeared in (\d{4})', dates)

print(match.group())
print(match.group(1))
print(match.group(2))
```

There is also `re.match()`, which is `re.search()` restricted to the *start* of the text. You will see it in other people's code; `re.search()` with a `^` anchor does the same and is clearer.

### `re.finditer()`: every match, with its position

`re.finditer()` is `re.findall()` with match objects: loop over it to get every match and where it is.

```python
for match in re.finditer(r'\d{4}', dates):
    print(match.group(), 'at position', match.start())
```

### `re.sub()`: replace what matches

`re.sub(pattern, replacement, text)` replaces **every** match of the pattern with the replacement string. It is `.replace()` for patterns, and the standard tool for cleaning text: strip out digits, collapse runs of whitespace, remove punctuation.

```python
messy = 'Chapter 12:   the   Queen\tof Hearts!!!   (page 87)'

print(re.sub(r'\d+', '#', messy))            # every number becomes #
print(re.sub(r'\s+', ' ', messy))            # any run of whitespace becomes one space
print(re.sub(r'[^\w\s]', '', messy))         # everything that is not a word character or whitespace is removed
print(re.sub(r'[^\w\s]', '', messy).lower().split())
```

The replacement can refer back to a group in the pattern with `\1`, `\2`, ...: this is how you *rearrange* text rather than delete it. Here we turn *Surname, Firstname* into *Firstname Surname*:

```python
names = 'Austen, Jane; Stoker, Bram; Eliot, George'

print(re.sub(r'(\w+), (\w+)', r'\2 \1', names))
```

### `re.split()`: split on a pattern

`text.split(', ')` splits on one fixed separator. `re.split(pattern, text)` splits wherever the pattern matches, which lets you split on *any* punctuation, on *any* amount of whitespace, or, roughly, into sentences:

```python
print(re.split(r'[,;] ?', 'Emma, Dracula;Middlemarch, Kidnapped'))
print(re.split(r'\s+', 'too   many    spaces\tand tabs'))

paragraph = 'Alice was tired. She looked at the book! Was it worth reading? It had no pictures.'
print(re.split(r'[.!?] ', paragraph))
```

This sentence splitter is already better than the `'. '` of the previous session, and still wrong for *Mr. Dodgson* or *etc.*; real sentence splitting needs more than a pattern, and you will meet tools for it later in the programme. For a bootcamp-level analysis, `re.split(r'[.!?]\s+', text)` is a fair approximation.

## Worked examples on a real text

Regular expressions earn their keep on a whole book. Some questions about *Alice* that string methods cannot answer in one line:

```python
from collections import Counter

with open('../data/alice.txt', encoding='utf-8') as infile:
    alice = infile.read()

# the chapter headings: CHAPTER followed by a space and Roman numerals
print(re.findall(r'CHAPTER [IVX]+', alice))
```

```python
# capitalised words, as whole words: names, mostly, plus sentence-initial words
capitalised = re.findall(r'\b[A-Z][a-z]+\b', alice)
print(len(capitalised))
print(Counter(capitalised).most_common(12))
```

```python
# words ending in -ing, lowercased, most common first
ing_words = re.findall(r'\b\w+ing\b', alice.lower())
print(Counter(ing_words).most_common(10))
```

```python
# tokenise with a pattern: letters and apostrophes, so that don't stays one word
tokens = re.findall(r"[a-z']+", alice.lower())
print(len(tokens))
print(tokens[:15])
```

Notice how each pattern was built: decide what a match looks like in words ("a capital, then lower-case letters, as a whole word"), translate it piece by piece, run it, look at the first results, adjust. Nobody writes a correct regex in one go; everybody tests. ## Common mistakes

Four regex errors that everyone makes.

```python
# Forgetting the raw string: Python interprets \b as a backspace before re ever sees it
print(re.findall('\bcat\b', 'the cat sat'))     # nothing found
print(re.findall(r'\bcat\b', 'the cat sat'))    # found
```

```python
# An unescaped full stop matches everything
print(re.findall(r'Mr.', 'Mr. Dodgson and Mrs Liddell and Mr Carroll'))
print(re.findall(r'Mr\.', 'Mr. Dodgson and Mrs Liddell and Mr Carroll'))
```

```python
# * means zero or more, so it also matches the empty string, everywhere
print(re.findall(r'\d*', 'a1b22'))
print(re.findall(r'\d+', 'a1b22'))
```

```python
# This cell produces an error on purpose: re.search() returned None because there was no match
match = re.search(r'\d{4}', 'no year here')
print(match.group())
```

## Quick reference

| Pattern | Matches |
| --- | --- |
| `abc` | exactly `abc` |
| `[abc]`, `[a-z]`, `[^abc]` | one of a, b, c; one in the range; one *not* a, b, c |
| `\d`, `\w`, `\s`, `.` | digit, word character, whitespace, any character |
| `\D`, `\W`, `\S` | not a digit, not a word character, not whitespace |
| `x+`, `x*`, `x?` | x one or more, zero or more, zero or one times |
| `x{3}`, `x{2,4}`, `x{2,}` | x three, two to four, two or more times |
| `^`, `$`, `\b` | start of text, end of text, word boundary |
| `a\|b`, `(...)`, `(?:...)` | a or b; a capturing group; a group that only groups |
| `\.`, `\?`, `\(` | the character itself |

| Function | Returns |
| --- | --- |
| `re.findall(p, t)` | list of all matches (or of the captured groups) |
| `re.search(p, t)` | first match as a match object, or `None` |
| `re.finditer(p, t)` | every match as a match object |
| `re.sub(p, r, t)` | `t` with every match replaced by `r` |
| `re.split(p, t)` | `t` split wherever `p` matches |

Add `re.IGNORECASE` as a last argument to ignore case.
