## Functions

You have been *using* functions since the first session: `print()`, `len()`, `int()`, `sorted()`, `open()`. A function is a piece of code with a name; you call it by the name, give it some input between the brackets, and get a result back. What you have not done yet is write your own. That is what this session is about, and it changes how you program: instead of one long stream of cells, you build a small set of named tools and combine them.

## From a block of code to a function

Here is an idiom you know well by now, a `for` loop with a test that filters a list. It keeps the five-letter words that start with *a*:

````python
words = ['abba', 'point', 'madam', 'feline', 'level', 'apple', 'google',
         'anton', 'microsoft', 'facebook', 'goggle', 'amazon', 'oracle', 'almost']

result = []
for word in words:
    if len(word) == 5 and word[0] == 'a':
        result.append(word)

print(result)
````

Fine for one list. But if you need the same filter on three lists, you copy the block three times, and when you find a mistake you fix it three times. The remedy is to give the block a **name** and make it reusable. That is a function definition:

```python
def name(parameters):
    body
    return result
```

- `def` is the keyword, followed by a **name** of your choice (same rules as for variables: `snake_case`, informative), then round brackets and a colon;
- between the brackets go the **parameters**: names for the input the function will receive;
- the **body** is indented, like the block of a loop or an `if`;
- `return` hands a value back to whoever called the function, and ends the function.

````python
def filter_words(word_list):
    result = []
    for word in word_list:
        if len(word) == 5 and word[0] == 'a':
            result.append(word)
    return result
````

Running that cell produces no output: it only *defines* the function. Nothing in the body is executed until you **call** the function, by its name, with a value between the brackets:

````python
print(filter_words(words))
````

What happens in a call: the value you pass, `words`, is assigned to the parameter `word_list`; the body runs with that; the `return` sends `result` back, and the call `filter_words(words)` *becomes* that value, exactly as `len(words)` becomes a number. You can call the function on any list, and store what comes back in a variable:

````python
short_list = filter_words(['albion', 'always', 'anton', 'anthony'])
print(short_list)

print(filter_words('a basket of fruit with one apple and one banana'.split()))
````

Two words you will hear all the time: the names in the definition (`word_list`) are the **parameters**; the actual values in a call (`words`) are the **arguments**. Parameters are variables that exist only while the function runs.

## `return` versus `print`

This is the point where most beginners stumble at least once, so read slowly. A function that **prints** shows something on the screen; a function that **returns** hands a value back to your code. They look similar when you try them in a notebook, and they are not the same thing at all.

````python
def shout_print(word):
    print(word.upper() + '!')

def shout_return(word):
    return word.upper() + '!'
````

````python
shout_print('hello')
shout_return('hello')
````

Both seem to work. Now try to *use* the result:

````python
a = shout_print('hello')
b = shout_return('hello')

print(a)
print(b)
````

`a` is `None`. **`None`** is Python's value for "nothing": a function without a `return` (or with a bare `return`) gives back `None`, and so does `print()` itself. `shout_print()` printed its text while it ran, but it handed nothing back, so there is nothing to store, compare, count or add to a list.

The rule: a function should **`return`** its result. Print inside a function only when printing *is* the job (a function that displays a table, say), and to inspect values while debugging. In the notebook, the value of the last line of a cell is shown automatically, which is why `shout_return('hello')` displayed something without a `print()`; in a script it would not.

`return` also ends the function immediately: any code after it is never reached. A function can have several `return` statements, in different branches, and the first one that executes wins.

````python
def describe_length(word):
    if len(word) > 8:
        return 'long'
    if len(word) > 4:
        return 'medium'
    return 'short'

print(describe_length('serendipitous'), describe_length('apple'), describe_length('a'))
````

## More than one parameter, and default values

Our filter is rigid: five letters and the letter *a* are fixed inside it. Make them parameters, and the function becomes a general tool:

````python
def filter_words(word_list, first_letter, length):
    result = []
    for word in word_list:
        if len(word) == length and word[0] == first_letter:
            result.append(word)
    return result

print(filter_words(words, 'a', 5))
print(filter_words(words, 'g', 6))
````

The arguments are matched to the parameters **by position**: the first goes to `word_list`, the second to `first_letter`, the third to `length`. Swap them and you get nonsense, and sometimes not even an error:

````python
print(filter_words(words, 6, 'g'))
````

That empty list is worse than an error: the function silently compared lengths to `'g'` and first letters to `6`, found nothing, and reported "no such words". We come back to this in a moment.

A parameter can have a **default value**, written `name=value` in the definition. If the call does not provide that argument, the default is used. Parameters with defaults must come after the ones without:

````python
def filter_words(word_list, first_letter='a', length=5):
    result = []
    for word in word_list:
        if len(word) == length and word[0] == first_letter:
            result.append(word)
    return result

print(filter_words(words))            # both defaults
print(filter_words(words, 'l'))       # length stays 5
print(filter_words(words, 'g', 6))
````

You can also name the arguments in the call: these are **keyword arguments**. They make the call readable, let you give the arguments in any order, and let you skip one default while setting the next:

````python
print(filter_words(words, length=6, first_letter='g'))
print(filter_words(words, length=6))    # first_letter keeps its default
````

You have used keyword arguments already: `print('a', 'b', sep='-')`, `open(path, encoding='utf-8')`, `sorted(numbers, reverse=True)`. Now you know how they are made.

### Checking the arguments

Back to the silent failure. A function that receives the wrong kind of input should say so, loudly, rather than compute nonsense. You know the tool from the session on `if` and `else`: `raise` an error with a clear message. `isinstance(value, type)` tests the type of a value.

````python
def filter_words(word_list, first_letter='a', length=5):
    if not isinstance(first_letter, str):
        raise TypeError(f'first_letter must be a string, got {first_letter}')
    if not isinstance(length, int):
        raise TypeError(f'length must be a whole number, got {length}')
    result = []
    for word in word_list:
        if len(word) == length and word[0] == first_letter:
            result.append(word)
    return result

print(filter_words(words, 'g', 6))
````

````python
# This cell produces an error on purpose: the arguments are swapped, and now the function says so
print(filter_words(words, 6, 'g'))
````

The checks sit at the top of the function, before any work is done, and the `raise` ends the function like a `return` would. `TypeError` is for the wrong type, `ValueError` for a value of the right type that makes no sense (a negative length, say). Two or three lines like these at the start of a function save hours of confusion later, for you and for whoever uses your function.

## Returning more than one value

A function can return only one thing, but that thing can be a tuple, and a tuple can hold as many values as you like. Write them after `return`, separated by commas, and unpack them where you call the function, the same way you unpack the pairs of `enumerate()` or `.items()`:

````python
def shortest_and_longest(word_list):
    shortest = word_list[0]
    longest = word_list[0]
    for word in word_list:
        if len(word) < len(shortest):
            shortest = word
        if len(word) > len(longest):
            longest = word
    return shortest, longest

short, long = shortest_and_longest(words)
print(short, long)

print(shortest_and_longest(['a', 'bb', 'ccc']))    # without unpacking: a tuple
````

## Docstrings and `help()`

A function's name says what it does; a **docstring** says it properly: a string on the first line of the body, in triple quotes, that describes what the function does, what it expects and what it returns. Python stores it, and `help()` shows it, exactly as it does for built-in functions. Write one for every function that is more than a few lines, or that someone else will use, including you in three weeks.

````python
def filter_words(word_list, first_letter='a', length=5):
    """Return the words in word_list that start with first_letter and have exactly `length` characters."""
    result = []
    for word in word_list:
        if len(word) == length and word[0] == first_letter:
            result.append(word)
    return result

help(filter_words)
````

## Inside and outside a function

Variables created inside a function, including its parameters, are **local**: they exist while the function runs and disappear when it returns. Outside the function they are unknown, and they do not interfere with variables of the same name outside. The technical term for this is **scope**: the part of a program in which a variable exists. A variable defined inside a function has the function as its scope; a variable defined in a cell, outside any function, is **global** and can be seen everywhere in the notebook.

````python
x = 12

def add_one(x):
    x = x + 1
    return x

print(add_one(5))
print(x)          # the outer x is untouched
````

````python
# This cell produces an error on purpose: total exists only inside the function
def total_length(word_list):
    total = 0
    for word in word_list:
        total += len(word)
    return total

print(total_length(words))
print(total)
````

A function *can* read variables from outside (the `vowels` string in the first class exercise, for instance), but it is cleaner to pass everything the function needs as an argument: the function is then self-contained and you can read it without looking elsewhere.

### One trap: lists

Numbers and strings are copied into a function in effect: whatever the function does with its parameter, the caller's variable is unchanged, as `x` was above. **Lists are different.** A parameter that receives a list is a second name for the *same* list, so if the function changes the list with `.append()`, `.sort()` or `del`, the caller's list changes too:

````python
def add_word(word_list):
    word_list.append('zebra')
    return word_list

animals = ['cat', 'dog']
more_animals = add_word(animals)

print(more_animals)
print(animals)        # changed as well: it is the same list
````

Sometimes that is what you want; usually it is a surprise. If a function should not touch the list it receives, work on a copy: `word_list.copy()` makes a new, independent list. The same holds for `b = a` outside functions: with lists, that gives two names for one list, not two lists.

````python
def add_word(word_list):
    new_list = word_list.copy()
    new_list.append('zebra')
    return new_list

animals = ['cat', 'dog']
more_animals = add_word(animals)

print(more_animals)
print(animals)
````

## Decomposition: small functions that call each other

The real power of functions shows when a problem is too big for one block of code. The strategy is called **decomposition**: split the problem into sub-problems, write a small function for each, and combine them. Each function does one thing, has a name that says what, and can be tested on its own. Functions can call other functions, including ones you wrote a minute ago.

Take a problem from the session on loops: for every word in a list that is longer than five letters, count its vowels. As one block:

````python
sentence = 'I like the words serendipitous and fictitious'
words = sentence.split()

result = []
vowels = 'aeiou'
for word in words:
    if len(word) > 5:
        vowel_count = 0
        for letter in word:
            if letter in vowels:
                vowel_count += 1
        result.append((word, vowel_count))

print(result)
````

It works, but it does three things at once. Decomposed, it becomes three functions: a test for a single letter, a count for a single word, and the loop over the list that uses the count.

````python
def is_vowel(letter):
    """Return True if letter is a vowel (lower or upper case)."""
    return letter.lower() in 'aeiou'

def count_vowels(word):
    """Return the number of vowels in word."""
    count = 0
    for letter in word:
        if is_vowel(letter):
            count += 1
    return count

def long_words_with_vowels(word_list, min_length=5):
    """Return (word, number of vowels) for every word longer than min_length."""
    result = []
    for word in word_list:
        if len(word) > min_length:
            result.append((word, count_vowels(word)))
    return result
````

````python
print(is_vowel('e'), is_vowel('X'))
print(count_vowels('abracadabra'))
print(long_words_with_vowels(words))
print(long_words_with_vowels(words, 3))
````

More lines than the single block, and much better code: every piece can be tested separately (`is_vowel('X')` is easy to check by eye), `count_vowels()` is reusable in any other program, and `min_length` became a parameter almost for free. This is how you will write anything larger than a few lines from now on, including the exam tasks, which often ask for exactly this kind of decomposition.

## Modules, once more

The session on files introduced `import module`, and the session on dictionaries `from module import name`. There is a third form, which you will use daily from the `pandas` session on: `import module as alias`, which loads a module under a shorter name.

````python
import random                      # random.randint(1, 6)
from random import randint         # randint(1, 6)
import random as rd                # rd.randint(1, 6)

print(random.randint(1, 6), randint(1, 6), rd.randint(1, 6))
````

All three do the same; they differ in what you type afterwards. Conventions exist for the aliases of the big libraries (`import pandas as pd`, `import numpy as np`), and you should follow them, because everyone else does.

A module is just a file of function definitions, which means your own functions can become a module: put them in a file `mytools.py` next to your notebook, and `import mytools` makes them available as `mytools.count_vowels()`. That is how code gets shared between notebooks and between people.

## Common mistakes

Four errors that everyone makes in their first week with functions.

````python
# Not a mistake that raises an error, which is what makes it nasty: the function is not *called*
def greet(name):
    return f'Hello, {name}!'

message = greet          # missing brackets: this is the function itself, not its result
print(message)
print(greet('Alice'))    # this is what was meant
````

````python
# Forgetting the return: the function computes the answer and then throws it away
def total_length(word_list):
    total = 0
    for word in word_list:
        total += len(word)

print(total_length(['a', 'bb']))    # None
````

````python
# This cell produces an error on purpose: the call gives two arguments to a function that takes one
def shout(word):
    return word.upper()

print(shout('hello', 'world'))
````

````python
# This cell produces an error on purpose: the function is called before it is defined (order matters)
print(whisper('HELLO'))

def whisper(word):
    return word.lower()
````
