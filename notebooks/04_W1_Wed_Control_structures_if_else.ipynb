## Control structures

This is the first session in a series on **control structures**, which tell the computer *when* to execute which piece of code. They are essential to programming, because without them a program would just be a list of instructions that run once, from top to bottom.

Control structures represent logical constructions that you know from everyday life:

1. **If** it is raining, take an umbrella. **Else**, don't bother.
2. **If** the traffic light is red, stop. **Else**, go. (But what about an amber light? We'll come to that.)

> **Tip: from prose to code, and back.** Translating code into prose (and prose into code) is a useful way of understanding what a program does. Python was designed to read easily, and as you improve, it will feel less and less like translating.

All of these examples use a similar structure: **test something, then do an action depending on the outcome of the test**. This is an important and recurring pattern in programming, because it is reliable and predictable.

## `if` and `else` in Python

To understand control structures in Python, we need three concepts: **keywords**, **indentation** and **logical expressions**. They are a big deal in Python, so take your time to practise them.

### Keywords

Keywords are a small set of reserved words that Python uses for specific actions; you saw the list in the session on variables, and you cannot use them as variable names. Today we use three of them: `if`, `else`, and their cousin `elif`.

### Logical expressions

A logical expression is a statement that evaluates to `True` or `False`. In short, it is a test. Some examples in prose are "Is the traffic light green?", "Is it raining?", or "Is this number larger than 5?".

### Indentation

Whitespace is the stuff that separates words: spaces, tabs, line breaks. In Python, whitespace at the *start* of a line has a meaning: it marks a **block of code** that belongs together. This is called **indentation**.

```python
a_number = 5

if a_number > 7:
    print('Hello World')
```

In the example, the four spaces before `print` separate that statement from the rest of the code: they say that the print statement belongs to the `if`, and is only executed when the test succeeds. The end of the indented block tells Python that the `if` statement has ended; whatever follows is executed regardless of the test.

Indentation is always four spaces per level. Jupyter inserts them for you when you press Enter after a colon, and pressing Tab inserts four spaces as well.

> **The beauty of Python: spaces hold meaning.** Most other programming languages do not treat whitespace this way; they use a keyword like `endif` or punctuation like `}` to end a block. Proponents of those languages say that Python is crazy. Python fans call it beautiful.

## `if`

We now combine these logical expressions with the `if` keyword. `if` involves a test and a conditional action: **if the test is true, then do something**. In Python, the test is a logical expression, like `3 > 4`, and the action is the indented block that follows.

> **Auto-complete.** After the `if` keyword and its test comes a colon (`:`). After that colon, Jupyter automatically indents the next line. If you forget the colon (it happens to the best of us), no indentation is added, which is a useful hint that something is not quite right.

An `if` without an indented block is not complete, and Python says so:

````python
# This cell produces an error on purpose: the if has no block
if 2 > 3:
````

We need an indented block with something to do:

````python
if 2 > 3:
    print('Am I going crazy?')
````

````python
if 2 > 1:
    print('All is fine!')
````

- In the first cell nothing was printed, because **the test result was `False`** (2 is not greater than 3): the condition was not met.
- In the second cell `All is fine!` was printed, because **the test result was `True`** (2 is greater than 1): the condition was met.

If you want to leave a block empty for the time being (because you will write it later), Python still insists on *something* there. The keyword `pass` does nothing, and exists for exactly this purpose:

````python
if 2 > 3:
    pass   # to be written later
````

## `else`

`else` comes after an `if` and cannot occur on its own. It is a *catch-all* clause: its block is executed whenever the `if` test was **not** `True`.

````python
traffic_light = 'green'

if traffic_light == 'red':
    print('stop')
else:
    print('go')
````

The `else` block is executed only when the `if` block is not. Exactly one of the two blocks runs.

## `elif`

`elif` is a blend of `else` and `if`:

- it always follows an `if` (it cannot be used on its own);
- it performs a further test, but only on the cases that the previous `if` (or `elif`) did *not* catch;
- it can be used multiple times; each `elif` tests the remainder of the previous tests;
- it is a cleaner way of writing the nested `if` you added above.

````python
traffic_light = 'red'

if traffic_light == 'red':
    print('stop')
elif traffic_light == 'amber':
    print('wait')
else:
    print('go')
````

We can use `input()` to make this a bit more dynamic. (You met `input()` in the exercises: it asks the user for an answer and returns it as a string.)

````python
traffic_light = input('What colour is the traffic light? ')

if traffic_light == 'red':
    print('stop')
elif traffic_light == 'amber':
    print('wait')
else:
    print('go')
````

### No overlap, and no loose ends

Good code is precise. An `if`-`elif`-`else` structure can introduce vagueness: each test should cover something the other tests do not, and the structure as a whole should deal with all possible cases.

**Class exercise:** the traffic-light code above is flawed. What happens to the car if there is a power cut and the light is off?

### Overlapping tests: `>=` and `<=`

`>=` and `<=` are a common source of unclear code in an `if`-`elif`-`else`, because they can make tests overlap:

````python
a_number = 4

if a_number <= 4:
    print('if-condition is met')
elif a_number >= 4:
    print('elif-condition is met')
````

Both logical expressions are `True` for the number 4, but only the `if` block runs, because it came first: once a test succeeds, the remaining `elif` tests are not even looked at. Avoid such overlaps by planning for all possible outcomes. As a rule of thumb, do not use more than one of `<=` or `>=` in the same `if`-`elif`-`else` structure.

## Combining tests: `and`, `or`, `not`

The key to using `if`, `elif` and `else` is effective testing. You have used `>`, `<`, `==` and `!=`; the keywords `and`, `or` and `not` from the session on data types let you combine tests into bigger ones.

### `or`

With `or`, the combined test is `True` if *at least one* of the two tests is `True`:

````python
traffic_light = 'amber'

if traffic_light == 'red' or traffic_light == 'amber':
    print('wait')
else:
    print('go')
````

Two things happened here: we removed the `elif`, and we tested two things in one test (is the light red? is the light amber?). Note that you have to write out both tests in full: `traffic_light == 'red' or 'amber'` is a classic mistake (it is always `True`, for reasons that we explain below).

### `and`

With `and`, the combined test is only `True` if *both* tests are `True`:

````python
temperature = 40

if temperature > 0 and temperature < 100:
    print("that's wet")
else:
    print("that's not wet")
````

### `not`

`not` inverts a test: `not x > 5` is `True` exactly when `x > 5` is `False`. It also gives us `not in`:

````python
word = 'rhythm'

if 'y' in word and 'e' not in word:
    print('a word with a y but without an e')
````

> **Note:** you will sometimes see `&` and `|` in Python code, for instance in the session on `pandas` later in the bootcamp. Those are different operators with a different purpose; in ordinary `if` tests, always use `and`, `or` and `not`.

## What counts as `True`?

You don't always need a comparison in a test: Python can treat *any* value as `True` or `False`. The rule is simple: zero, an empty string, an empty list and `None` count as `False`; everything else counts as `True`. This is often used to test whether something is empty:

````python
name = ''

if name:
    print(f'Hello, {name}!')
else:
    print('You did not enter a name.')
````

````python
words = ['some', 'words']

if words:
    print('the list is not empty')
if not []:
    print('an empty list counts as False')
````

(This is also why `traffic_light == 'red' or 'amber'` is always `True`: the string `'amber'` is not empty, so it counts as `True` on its own.)

## A few more things about tests

**A test is a value.** A comparison produces a boolean, and you can store it in a variable and use the variable in an `if`. This often makes code easier to read:

````python
word = 'incomprehensible'
is_long = len(word) > 10
print(is_long)

if is_long:
    print('a long word')
````

**Brackets when mixing `and` and `or`.** Python evaluates `and` before `or`, so `a or b and c` means `a or (b and c)`, which is rarely what you intended. When you mix the two, put brackets around the parts, so that the reader (and your future self) can see what belongs together:

````python
is_weekend = True
is_holiday = False
is_raining = True

if (is_weekend or is_holiday) and not is_raining:
    print('picnic')
else:
    print('reading')
````
