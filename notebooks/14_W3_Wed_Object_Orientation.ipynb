## Objects, Plato and cookies

Python is what people call an **object-oriented** language: **objects** are central to how you program in it. In fact you have worked with objects all along, because *everything* in Python is an object, down to the humblest integer or string. This session makes explicit what you have been doing for three weeks, so that you understand the words in error messages and documentation (*object*, *class*, *instance*, *attribute*, *method*), and, at the end, shows you what a class looks like from the inside. You will not be asked to write classes at the exam; you will be reading and using them for the rest of your programming life.

After this session you can:

- explain what a class and an instance are, and what `type()` tells you;
- tell an attribute from a method, and a method from a function;
- explain why some methods change the object they are called on and others return a new one;
- read a simple class definition: `class`, `__init__`, `self`, methods, `__str__`.

To understand objects, it helps to turn to Plato. Remember his theory of ideas: for every kind of thing in the world there is an *ideal* form, perfect and never actually observed, of which all the real things are imperfect copies. Whenever I think of Plato, I think of cookies. There is something like the ideal chocolate chip cookie, the mould from which all cookies are cut; and then there are the actual cookies, each slightly different from the next and from the ideal.

Now think of baking cookies with Plato, and run this line:

```python
from collections import Counter

cnt = Counter()
```

Running that line, that is, calling the **constructor**, is like cutting one cookie with the `Counter` mould. The constructor hands you an actual cookie, made after the abstract `Counter` blueprint that lives in the `collections` module. As with real cookies, you can cut as many independent ones as you like from the same mould, and the mould stays what it is:

```python
cnt1 = Counter()
cnt2 = Counter()
cnt3 = Counter()

cnt1.update('aaa')
print(cnt1)
print(cnt2)      # untouched: a different cookie
```

In Python the abstract template is called a **class**, and every cookie cut from it an **instance** (or *object*). By convention, class names are written with a capital (`Counter`, `DataFrame`) and instances in lower case (`cnt`, `df`). When you ask for the `type()` of something, you are really asking: *from which mould was this cookie cut?*

```python
print(type(cnt3))
print(type('hello'), type(42), type([1, 2]), type({'a': 1}))
```

So `str`, `int`, `list` and `dict` are classes too, and every string you ever typed was an instance of `str`. The same goes for the DataFrame of the `pandas` session: `pd.DataFrame` is a class, and `pd.read_csv()` is a function that constructs an instance of it for you.

A constructor can take arguments that set the new object up. `Counter` accepts anything you can loop over, and the new instance starts out holding the counts:

```python
aeneid = """Arma virumque cano, Troiae qui primus ab oris
Italiam fato profugus Lavinaque venit
litora multum ille et terris iactatus et alto
vi superum, saevae memorem Iunonis ob iram,
multa quoque et bello passus, dum conderet urbem
inferretque deos Latio; genus unde Latinum
Albanique patres atque altae moenia Romae."""

letter_counts = Counter(aeneid)
print(letter_counts.most_common(5))
```

### Attributes and methods

Cookies have properties (a cookie is sweet, it has twelve chocolate chips) and things they can do or that can be done to them (it can be baked, or eaten). Objects are the same: they carry **attributes**, values stored on the object, and **methods**, functions that belong to the object. Both are reached with the **dot**. You saw both in the `pandas` session without naming them: `df.shape` is an attribute (a value; no brackets), `df.head()` is a method (a function; brackets, and possibly arguments).

```python
import pandas as pd

df = pd.read_csv('../data/screenplays.csv')

print(df.shape)              # attribute: a stored value, no brackets
print(df.columns[:3])        # attribute
print(df.head(2))            # method: something the object does, with brackets
```

## Methods versus functions

Pay attention to the language: we say that we call a method **on** an object, `cnt.most_common()`, rather than passing the object to a function, `len(cnt)`. The distinction matters, and sorting a list shows why. First we make a list, by calling the constructor of the `list` class on a string:

```python
letters = list('platonic')
print(letters)
print(type(letters))
```

To sort it there are two options. The first is the general **function** `sorted()`, to which you pass the list. It returns a sorted **copy** and leaves the original alone:

```python
sorted_letters = sorted(letters)
print(sorted_letters)
print(letters)           # unchanged
```

`sorted()` is not tied to lists: it accepts anything you can loop over, a string included, and always returns a list. The second option is the **method** `.sort()`, called *on* the list itself:

```python
letters.sort()
```

Nothing was printed, and that is normal: `.sort()` sorts the list **in place** and returns nothing, that is, `None`:

```python
print(letters.sort())    # None: the method does not hand anything back
print(letters)           # but the list is sorted now
```

`.sort()` can do this because it is not just any function: it belongs to the list, it was part of the mould, and it has privileged access to the object's insides. That is what a method is: a function tied to a class, which every instance of that class carries around. Two consequences:

- all methods are functions, but not all functions are methods;
- a method exists only for the objects of its class: `'abc'.sort()` fails, because strings have no such method (and could not have one: a string cannot be changed in place, as you saw in the session on strings).

This explains three surprises from the past weeks in one go. `text.lower()` and `df.sort_values()` return a **new** object and leave the original unchanged, because strings and DataFrames are not changed in place; `.append()` and `.sort()` change the list in place and return `None`, so `words = words.sort()` throws your list away. When in doubt, `help()` says which of the two a method does.

```python
# This cell produces an error on purpose: strings have no sort() method
'abc'.sort()
```

## Home-baked cookies

This last part is not something you need to master now: you will not write classes for a while, simply because you will not feel the need, with so many ready-made ones to import. But reading a class definition once makes everything above concrete, and you will meet class definitions in other people's code soon enough.

We **reverse-engineer** a lightweight `Counter`. The first thing a class needs is a name and a constructor. The syntax: the keyword `class`, a name, a colon, and an indented body of method definitions. The constructor is a method with the special name `__init__` (two underscores on each side, pronounced *dunder init*), which Python calls whenever a new instance is made:

```python
class OurCounter:

    def __init__(self):
        pass
```

```python
cnt = OurCounter()
print(type(cnt))
```

That is a class that does nothing, but it already shows the three things to notice:

- the constructor looks like a normal function definition, indented inside the class, which makes it a **method** of `OurCounter`;
- its name `__init__` is fixed, so that Python knows to call it at construction;
- its first parameter is **`self`**: the object itself. Every method receives the object it is called on as its first argument, and `self` is the conventional name for it.

A real `Counter` is a dictionary with extras. So the first thing our constructor should do is give every new instance a dictionary of its own. We store it *on the object*, with `self.`:

```python
class OurCounter:

    def __init__(self):
        self.counts = {}


cnt = OurCounter()
print(cnt.counts)
```

`self.counts = {}` creates an **attribute**: the same dot syntax you use from the outside (`cnt.counts`) used from the inside, where the object calls itself `self`. Two instances get two separate dictionaries, because `__init__` runs once per cookie.

Now a method. Counting means: add one to the count of a key, or start its count if the key is new. That is the `try`/`except` from the session on dictionaries, wrapped in a method, with the amount as a parameter with a default:

```python
class OurCounter:

    def __init__(self):
        self.counts = {}

    def add(self, key, amount=1):
        try:
            self.counts[key] += amount
        except KeyError:
            self.counts[key] = amount


cnt = OurCounter()
cnt.add('a', 2)
cnt.add('b', 3)
cnt.add('a', 5)
cnt.add('z')
print(cnt.counts)
```

Notice the two views. From the outside, the user writes `cnt.add('a', 2)`: two arguments. Inside, the method has three parameters, because `self` comes first and Python fills it in with `cnt`. And inside, the dictionary is `self.counts`, the object's own. The user never has to know there is a dictionary at all: that hiding of the messy inside behind a clean outside is called **abstraction**, and it is the point of writing classes.

A `most_common()` method uses the sort-by-value trick from the session on comprehensions:

```python
class OurCounter:

    def __init__(self):
        self.counts = {}

    def add(self, key, amount=1):
        try:
            self.counts[key] += amount
        except KeyError:
            self.counts[key] = amount

    def most_common(self, n=None):
        pairs = sorted(self.counts.items(), key=lambda pair: pair[1], reverse=True)
        return pairs[:n]


cnt = OurCounter()
cnt.add('a', 1)
cnt.add('b', 3)
cnt.add('c', 15)
cnt.add('d', 4)
print(cnt.most_common(2))
print(cnt.most_common())
```

(`pairs[:None]` is the whole list: slicing with `None` as the end means "to the end", which is how `n=None` gives *everything* when no number is passed.)

The real `Counter` can be filled at construction, `Counter(aeneid)`. We give our constructor an optional argument and let it call our own `add()` method, once per item. Yes: inside the constructor you can already use the methods defined further down.

```python
class OurCounter:

    def __init__(self, iterable=None):
        self.counts = {}
        if iterable:
            for item in iterable:
                self.add(item)

    def add(self, key, amount=1):
        try:
            self.counts[key] += amount
        except KeyError:
            self.counts[key] = amount

    def most_common(self, n=None):
        pairs = sorted(self.counts.items(), key=lambda pair: pair[1], reverse=True)
        return pairs[:n]


cnt = OurCounter(aeneid)
print(cnt.most_common(5))
print(Counter(aeneid).most_common(5))     # the real one agrees
```

One last touch. Print the real `Counter` and you get a readable line; print ours and you get its address in memory:

```python
print(Counter('abca'))
print(OurCounter('abca'))
```

Every object has a method `__str__` that Python calls when the object is printed; the default one is that unhelpful address. A class can **override** it by defining its own, returning the string it wants to show. The double underscores mark it, like `__init__`, as one of the methods that Python itself calls behind the scenes:

```python
class OurCounter:

    def __init__(self, iterable=None):
        self.counts = {}
        if iterable:
            for item in iterable:
                self.add(item)

    def add(self, key, amount=1):
        try:
            self.counts[key] += amount
        except KeyError:
            self.counts[key] = amount

    def most_common(self, n=None):
        pairs = sorted(self.counts.items(), key=lambda pair: pair[1], reverse=True)
        return pairs[:n]

    def __str__(self):
        return 'OurCounter(' + str(self.counts) + ')'


cnt = OurCounter('abca')
print(cnt)
```

That is a class: a name, a constructor that sets up attributes on `self`, methods that use and change them, and the occasional dunder method that plugs into Python's own machinery. You have now seen the whole mechanism behind every `.something()` you have typed since the first session.
